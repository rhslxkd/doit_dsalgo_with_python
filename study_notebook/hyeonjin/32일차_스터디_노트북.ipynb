**이 트리 하나로 1~10번을 전부 풉니다.** 종이에 크게 옮겨 그려두면 편해.

```
                        [ A ]                        ← 레벨 0
                    ／    |    ＼
                  ／      |      ＼
              [ B ]     [ C ]     [ D ]              ← 레벨 1
             ／   ＼             ／   ＼
         [ E ]   [ F ]       [ G ]   [ H ]           ← 레벨 2
        ／  |  ＼                    ／   ＼
    [ I ][ J ][ K ]              [ L ]   [ M ]       ← 레벨 3
```

**연결 관계 정리** (헷갈리면 여기 보고 확인)
- A의 자식: B, C, D
- B의 자식: E, F
- C의 자식: 없음
- D의 자식: G, H
- E의 자식: I, J, K
- F, G의 자식: 없음
- H의 자식: L, M
- I, J, K, L, M의 자식: 없음

**연결 관계** (반드시 확인하고 시작해)

| 노드 | 왼쪽 자식 | 오른쪽 자식 |
|---|---|---|
| A | B | C |
| B | D | E |
| C | F | G |
| D | **H** | **없음** ⚠️ |
| E | I | J |
| F | K | L |
| G | **없음** | **없음** |
| H, I, J, K, L | 없음 | 없음 |

🔥 **함정 두 개를 미리 짚어둘게:**
- `D` 는 **왼쪽 자식만** 있어 (H). 오른쪽은 비었지.
- `G` 는 **리프**야. 레벨 2인데 자식이 없어.

이 두 가지 때문에 순회 결과가 헷갈리기 쉬워. 표를 옆에 두고 풀어.

### 16. 🔴 [설명] 🔥 "지나가는 것"과 "방문하는 것"은 다르다

**오늘 가장 중요한 개념이야.** 교재 380p:

> **"[그림 9-5]는 스캔 과정에서 노드 A를 몇 번 지나갔는지를 보여 줍니다. 다음과 같이 깊이 우선 검색을 하면서 노드 A를 3번 지나갑니다."**
> ① A에서 B로 내려가기 직전에 A를 1번 지나갑니다.
> ② B에서 C로 지나가는 도중에 A를 1번 지나갑니다.
> ③ C에서 A로 돌아올 때 A를 1번 지나갑니다.

그리고 결정적으로:
> **"이때 어느 시점에 실제로 노드를 방문하는지에 따라 깊이 우선 검색은 [그림 9-5]와 같이 세 종류의 스캔 방법으로 구분합니다."**
> **"지나가는 것과 방문하는 것은 다른 개념입니다."**

**이 그림을 머릿속에 새겨:**
```
                    A
                  ／ ｜ ＼
              ①  ／  ｜②  ＼ ③
                ／   ｜    ＼
              B      ↓       C
                  (되돌아옴)

  ① 첫 번째 지나감 → 여기서 방문하면 전위 순회
  ② 두 번째 지나감 → 여기서 방문하면 중위 순회
  ③ 세 번째 지나감 → 여기서 방문하면 후위 순회
```

**질문에 답해봐:**
- 노드 하나를 최대 **몇 번** 지나가? ①____
- 자식이 **하나뿐**인 노드(예: `D`)는 몇 번 지나갈까? ②____ 왜? ③________________
- **리프**(예: `H`)는 몇 번 지나갈까? ④____
- 🔥 **"지나감"과 "방문"의 차이를 한 문장으로:** ⑤________________
- 세 가지 순회는 결국 **"같은 경로를 걷되, 언제 이름을 적느냐"** 의 차이야. 경로 자체는 ⑥________ (같다/다르다)

💡 이 개념을 잡으면 세 순회를 **외울 필요가 없어져.** 경로는 하나고, 기록 시점만 셋이니까.

### 14. 필요한 자료구조

**큐 추적**

| 단계 | 꺼낸 노드 | 넣은 것 | 큐 상태 |
|---|---|---|---|
| 1 | A | ④ **B, C** | ⑤ `[B, C]` |
| 2 | B | ⑥ **D, E** | ⑦ `[C, D, E]` |
| 3 | C | ⑧ **F, G** | ⑨ `[D, E, F, G]` |
| 4 | D | ⑩ **H** | ⑪ `[E, F, G, H]` |
| 5 | E | ⑫ **I, J** | ⑬ `[F, G, H, I, J]` |

> 🔑 **자료구조가 순회 순서를 결정한다.** 큐(FIFO) → 너비 우선, 스택(LIFO) → 깊이 우선. 4일차에 배운 두 구조가 여기서 만나.

- ① **한 가지를 리프까지 끝까지 파고든 뒤 되돌아온다**
- ② **백트래킹(backtracking)** / 되돌아가기
- ③ `H` 에서 **`D` 로 되돌아감** ④ `D` 의 오른쪽 자식이 없으니 **`B` 로 되돌아가서 `E` 로 내려감**

### 16. 🔥 "지나감" vs "방문"
- ① **3번** ② **2번** ③ **자식이 하나면 "자식 사이를 지나가는" 두 번째 통과가 없으니까**
- ④ **1번** (리프는 내려갈 곳이 없어 한 번만 스침)
- ⑤ **"지나감"은 경로상 그 노드를 거치는 것이고, "방문"은 그때 실제로 데이터를 처리(출력/기록)하는 것.**
- ⑥ **같다** 🔥

> 🔑 **세 순회는 같은 길을 걷는다. 다른 건 "언제 이름을 적느냐"뿐이다.**
> 이걸 잡으면 세 순회를 따로 외울 필요가 없어져. 경로를 한 번 그리고 ①②③ 중 어디서 적을지만 정하면 되니까.

### 17. 전위 순회
```
전위(노드):
    ① 노드를 방문한다
    ② 왼쪽 자식에 대해 전위 순회
    ③ 오른쪽 자식에 대해 전위 순회
```

**전체 순서** ④:
```
A → B → D → H → E → I → J → C → F → K → L → G
```
(교재 381p와 일치 ✅)

- ⑤ **루트** ⑥ **어떤 트리든 성립해** (노드를 가장 먼저 방문하니까)
- ⑦ **가장 오른쪽 경로의 끝** — 여기서는 `G`

### 18. 중위 순회
```
중위(노드):
    ① 왼쪽 자식에 대해 중위 순회
    ② 노드를 방문한다
    ③ 오른쪽 자식에 대해 중위 순회
```

**전체 순서** ④:
```
H → D → B → I → E → J → A → K → F → L → C → G
```
(교재 381p와 일치 ✅)

⚠️ **`D` 를 주의해야 했지**: 왼쪽 `H` → `D` 방문 → 오른쪽 없음. 그래서 `H, D` 순.

- ⑤ **7번째** (12개 중 정확히 가운데는 아니지만, 왼쪽 서브트리 6개 다음)
- ⑥ **왼쪽 서브트리 전체** (H, D, B, I, E, J) ⑦ **오른쪽 서브트리 전체** (K, F, L, C, G)

> 🔑 **중위 순회에서 루트는 왼쪽 서브트리와 오른쪽 서브트리를 정확히 가르는 경계**야. 이게 22번 트리 복원의 핵심이야.

### 19. 후위 순회
```
후위(노드):
    ① 왼쪽 자식에 대해 후위 순회
    ② 오른쪽 자식에 대해 후위 순회
    ③ 노드를 방문한다
```

**전체 순서** ④:
```
H → D → I → J → E → B → K → L → F → G → C → A
```
(교재 381p와 일치 ✅)

- ⑤ **루트** (항상 맨 마지막)
- ⑥ **가장 왼쪽 경로의 끝** — 여기서는 `H`
- ⑦ **같아!** 리프는 세 순회 모두 `H → I → J → K → L → G` 순서를 유지해.
  🔥 **왜?** 리프는 자식이 없어서 "노드 방문"이 유일한 이벤트거든. 순서를 바꿀 여지가 없지.

### 20. 세 순회 비교

| | 순서 규칙 | 루트 위치 | 첫 노드 | 마지막 노드 |
|---|---|---|---|---|
| 전위 | ① 노드→왼→오 | ② **맨 앞** | ③ **A** | ④ **G** |
| 중위 | ⑤ **왼→노드→오** | ⑥ **가운데** | ⑦ **H** | ⑧ **G** |
| 후위 | ⑨ **왼→오→노드** | ⑩ **맨 뒤** | ⑪ **H** | ⑫ **A** |

- ⑬ **먼저(pre)** ⑭ **사이에(in)** ⑮ **나중에(post)**

> 🔑 **왼쪽이 오른쪽보다 먼저인 건 세 순회 모두 같아.** 오직 **노드 자신의 위치**만 앞/중간/뒤로 바뀌지. 그래서 이름이 pre/in/post인 거고.

### 21. 작은 트리 연습

**(가)**
| | 순서 |
|---|---|
| 전위 | ① **A B C** |
| 중위 | ② **B A C** |
| 후위 | ③ **B C A** |

**(나)**
| | 순서 |
|---|---|
| 전위 | ④ **A B D C E** |
| 중위 | ⑤ **D B A C E** |
| 후위 | ⑥ **D B E C A** |

**(다)** ⚠️ `B` 의 오른쪽이 `C`, `C` 의 왼쪽이 `D`
| | 순서 |
|---|---|
| 전위 | ⑦ **A B C D** |
| 중위 | ⑧ **B D C A** |
| 후위 | ⑨ **D C B A** |

🔥 **(다) 해설**: 전위는 `A → B(왼쪽) → B의 왼쪽 없음 → B의 오른쪽 C → C의 왼쪽 D`. 중위는 `B의 왼쪽 없음 → B → C의 왼쪽 D → C → C의 오른쪽 없음` 하고 `A`. 후위는 자식부터니까 `D → C → B → A`.

## 🧠 PART 5 해설

### 22. 트리 복원

**문제 (가)**
- ① **P** ② **S Q T** ③ **U R V**
- ④ **Q S T** ⑤ **R U V**
- ⑥ **Q** ⑦ **S** ⑧ **T**
- ⑨ **R** ⑩ **U** ⑪ **V**

**완성된 트리** ⑫:
```
            P
         ／     ＼
      Q           R
    ／  ＼      ／  ＼
   S     T    U      V
```

**검산** ⑬ 후위 순회: **S T Q U V R P** ✅

**문제 (나)**
- ① **P** ② **Q, S** ③ **R, T, U**
- 트리:
```
        P
      ／   ＼
    Q        R
      ＼    ／
       S   T
          ／
         U
```
(`Q` 의 **오른쪽** 자식이 `S`, `R` 의 **왼쪽** 자식이 `T`, `T` 의 **왼쪽** 자식이 `U`)
- ④ 후위: **S Q U T R P** ✅

**문제 (다)** 🔥
- **노드 2개**: ⑤ **둘 다 가능!**
  - ⑥ B가 왼쪽 → 중위 **B A**
  - ⑦ B가 오른쪽 → 중위 **A B**
- **노드 3개**: ⑧ **4개**
  - ⑨ 각각의 중위: **ABC, ACB, BCA, CBA**
```
  1) A→B(오)→C(오)   중위 ABC
  2) A→B(오), B→C(왼) 중위 ACB
  3) A→B(왼), B→C(오) 중위 BCA
  4) A→B(왼)→C(왼)   중위 CBA
```
- ⑩ **전위와 후위는 둘 다 "루트가 어디인가"만 알려주고, "왼쪽 서브트리가 어디까지인가"를 알려주지 못하기 때문.** 자식이 하나뿐인 노드에서 그게 왼쪽인지 오른쪽인지 판별할 정보가 없어.
- ⑪ **중위**
- ⑫ **정 이진 트리는 자식이 0개 아니면 2개**라서, 자식이 있으면 반드시 좌우 둘 다 있어. "어느 쪽인가"의 모호함이 애초에 생기지 않지.

> 🔑 **전위 + 중위 = 복원 가능, 후위 + 중위 = 복원 가능, 전위 + 후위 = 일반적으로 불가능.**
> **중위 순회가 좌우 경계 정보를 유일하게 제공**하기 때문이야.

### 23. 수식 트리
- ① 전위: **`* + 3 5 - 10 4`**
- ② 중위: **`3 + 5 * 10 - 4`**
- ③ 후위: **`3 5 + 10 4 - *`**
- ④ **전위(prefix)** ⑤ **중위(infix)** ⑥ **후위(postfix)**

🔥 **중위 표기의 문제**
- ⑦ `3 + 5 * 10 - 4` 를 그냥 계산 = **49** (곱셈 우선이라 `3 + 50 - 4`)
- ⑧ 원래 트리의 정답 = **48** (`8 × 6`)
- ⑨ **트리 구조가 담고 있던 "먼저 계산할 순서"가 평평한 문자열로 바뀌면서 사라졌기 때문.** `+` 가 `*` 보다 먼저 계산되어야 한다는 정보를 표기 자체가 담지 못해.
- ⑩ **괄호** — `(3 + 5) * (10 - 4)`

🔥 **후위 표기의 강점**
- ⑪ **없어**
- ⑫ **연산자가 나오는 순간 "바로 앞의 두 값"에 적용된다는 규칙이 순서를 유일하게 결정하기 때문.** 우선순위도 괄호도 필요 없어.
- ⑬ **스택에 push** ⑭ **스택에서 2개를 pop해서 연산하고 결과를 다시 push**

**직접 계산해보면**:
```
3      → [3]
5      → [3, 5]
+      → [8]           (3+5)
10     → [8, 10]
4      → [8, 10, 4]
-      → [8, 6]        (10-4)
*      → [48]          (8*6)  ✅
```

💡 실제 계산기와 컴파일러가 이 방식을 써. 4일차 스택이 여기서 쓰이는 거지.

### 24. 각 순회의 용도

| 상황 | 순회 | 이유 |
|---|---|---|
| 폴더 구조 출력 | ① **전위** | ② 상위 폴더를 먼저 보여줘야 계층이 보임 |
| 폴더 용량 계산 | ③ **후위** 🔥 | ④ 하위 폴더 용량을 모두 알아야 자신의 용량을 계산 가능 |
| BST 오름차순 출력 | ⑤ **중위** | ⑥ 왼쪽(작은 값) → 자신 → 오른쪽(큰 값) 순 |
| 트리 삭제 | ⑦ **후위** | ⑧ 자식을 먼저 지워야 안전 (부모를 먼저 지우면 자식에 접근 불가) |
| 트리 복사 | ⑨ **전위** | ⑩ 부모를 먼저 만들어야 자식을 붙일 수 있음 |

**핵심 판단 기준**
- ⑪ **후위** ⑫ **전위** ⑬ **중위**

**폴더 용량 예시**
- ⑭ `a.py`(3KB)와 `b.py`(5KB)를 먼저 알아야 → `src` = 8KB
- ⑮ `src`(8KB)와 `docs`(2KB)를 먼저 알아야 → `프로젝트` = 10KB
- → **계산이 아래에서 위로 올라와야 하니 후위 순회** ✅

> 🔑 **"자식의 결과가 필요하면 후위, 부모의 정보가 필요하면 전위"** — 이 한 줄이면 실전에서 헷갈릴 일이 없어.

### 25. 09-2 예고 — 이진 검색 트리
- ① **맞아** (20 < 30 < 40) ② **맞아** (20, 30, 40 < 50)
- ③ 중위 순회: **20 → 30 → 40 → 50 → 60 → 70 → 80**
- ④ 🔥 **오름차순으로 정렬되어 나온다!**
- ⑤ **더 작은 값** ⑥ **더 큰 값**
- ⑦ **가장 왼쪽 끝 노드** (왼쪽으로만 계속 내려가면 나옴) — 여기서는 `20`
- ⑧ **가장 오른쪽 끝 노드** — 여기서는 `80`
- ⑨ **찾는 값이 노드보다 작으면 왼쪽, 크면 오른쪽으로만 내려가면 돼.** 한 번 비교할 때마다 후보가 절반씩 줄어드니 **12일차 이진 검색과 똑같은 O(log n)** 이야.
  단, 트리가 한쪽으로 치우치면(8번의 "높이" 참고) O(n)이 될 수 있어 — 그게 09-2의 주요 논점이 될 거야.

## 📌 핵심 3줄 요약

1. **트리 용어는 "기준이 무엇인가"로 갈린다.** 리프는 *레벨*이 아니라 *자식 유무*, 형제는 *레벨*이 아니라 *부모*, 차수는 *자식의 수*, 높이는 *리프 레벨의 최댓값*. 이 기준만 정확하면 어떤 트리를 봐도 용어를 짚어낼 수 있어.
2. **세 가지 순회는 같은 길을 걷고 기록 시점만 다르다.** 깊이 우선으로 걸으면 각 노드를 최대 3번 지나가는데, 1번째에 적으면 전위, 2번째면 중위, 3번째면 후위야. 외울 게 아니라 **하나의 경로에서 파생되는 셋**이지.
3. **자료구조가 순회 순서를 결정한다.** 큐(FIFO)를 쓰면 너비 우선, 스택(LIFO)을 쓰면 깊이 우선. 그리고 **중위 순회만이 좌우 경계 정보를 담기 때문에** 전위+중위로는 트리를 복원할 수 있지만 전위+후위로는 못 해.
