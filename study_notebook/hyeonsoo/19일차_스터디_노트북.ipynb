## 📦 부록 — 정렬 함수 모음

이 셀을 **먼저 실행**해. (비교/교환 횟수를 셀 수 있는 버전)

````
from typing import MutableSequence
import random, bisect

# ---------- 18일차: 버블 정렬 ----------
def bubble_sort(a, count=False):
    c = s = 0
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):
            c += 1
            if a[j-1] > a[j]:
                s += 1
                a[j-1], a[j] = a[j], a[j-1]
    return (c, s) if count else a

# ---------- 실습 6-6: 단순 선택 정렬 ----------
def selection_sort(a, count=False):
    c = s = 0
    n = len(a)
    for i in range(n - 1):
        m = i                          # 최솟값의 '인덱스'
        for j in range(i + 1, n):
            c += 1
            if a[j] < a[m]:
                m = j
        if m != i:
            s += 1
        a[i], a[m] = a[m], a[i]        # 패스당 교환 딱 1번
    return (c, s) if count else a

# ---------- 실습 6-7: 단순 삽입 정렬 ----------
def insertion_sort(a, count=False):
    c = s = 0
    n = len(a)
    for i in range(1, n):
        j = i
        tmp = a[i]                     # 값을 '손에 쥠'
        while j > 0 and a[j-1] > tmp:
            c += 1
            a[j] = a[j-1]              # 이웃을 오른쪽으로 밀기
            s += 1
            j -= 1
        if j > 0:
            c += 1                     # while 조건이 거짓이 된 마지막 비교
        a[j] = tmp                     # 빈자리에 삽입
    return (c, s) if count else a

# ---------- 실습 6C-1: 이진 삽입 정렬 ----------
def binary_insertion_sort(a):
    n = len(a)
    for i in range(1, n):
        key = a[i]
        pl, pr = 0, i - 1
        while True:
            pc = (pl + pr) // 2
            if a[pc] == key:
                break
            elif a[pc] < key:
                pl = pc + 1
            else:
                pr = pc - 1
            if pl > pr:
                break
        pd = pc + 1 if pl <= pr else pr + 1
        for j in range(i, pd, -1):
            a[j] = a[j-1]
        a[pd] = key
    return a

# ---------- 실습 6C-2: bisect 사용 ----------
def binary_insort(a):
    for i in range(1, len(a)):
        bisect.insort(a, a.pop(i), 0, i)
    return a


def test_sort(f, trials=300):
    fails, example = 0, None
    for _ in range(trials):
        x = [random.randint(0, 50) for _ in range(random.randint(0, 10))]
        expected = sorted(x)
        y = x[:]
        f(y)
        if y != expected:
            fails += 1
            if example is None:
                example = (x, y, expected)
    return fails, example

print('준비 완료')
````

## 📖 오늘의 핵심 개념

### 1. 단순 선택 정렬 (straight selection sort)
> **아직 정렬 안 된 부분에서 가장 작은 원소를 "선택"해서, 그 부분의 맨 앞과 교환**

```python
for i in range(n - 1):
    min = i                          # 최솟값의 '인덱스' (값이 아님!)
    for j in range(i + 1, n):
        if a[j] < a[min]:
            min = j
    a[i], a[min] = a[min], a[i]      # 다 훑은 뒤 딱 한 번 교환
```

**버블과의 결정적 차이**:
- 버블: "비교하다 조건 맞으면 **즉시** 교환" → 패스당 교환 여러 번
- 선택: "끝까지 훑어서 **위치만 기억**했다가 마지막에 한 번" → 패스당 교환 **정확히 1번**

**비교 횟수**: `(n² - n) / 2` — 버블과 동일
**교환 횟수**: 최대 `n - 1`번 — 버블(`n(n-1)/2`)보다 훨씬 적음 ⭐

### 2. ⚠️ 선택 정렬은 안정적이지 않다
교재 239p: 값이 같은 원소 두 개(`3ᴸ`, `3ᴿ`)의 순서가 정렬 후 **뒤바뀔 수 있어**.

**왜?** 선택 정렬은 **멀리 떨어진 원소끼리 교환**해. 교환 한 번으로 중간의 다른 원소들을 **건너뛰어** 자리를 옮기니, 같은 값끼리의 원래 순서가 깨질 수 있어.

### 3. 단순 삽입 정렬 (straight insertion sort)
> **주목한 원소를, 이미 정렬된 앞부분의 알맞은 위치에 "삽입"**

```python
for i in range(1, n):
    j = i
    tmp = a[i]                        # ① 값을 손에 쥠 (자리는 빈 구멍이 됨)
    while j > 0 and a[j-1] > tmp:
        a[j] = a[j-1]                 # ② 이웃을 구멍 쪽으로 밀기
        j -= 1                        #    구멍이 왼쪽으로 이동
    a[j] = tmp                        # ③ 손에 쥔 값을 구멍에 꽂기
```

**⚠️ 헷갈리기 쉬운 지점**: `a[j] = a[j-1]`은 **스왑이 아니야.**
- 버블 정렬: `a[j-1], a[j] = a[j], a[j-1]` — 두 값이 **서로** 자리를 맞바꿈
- 삽입 정렬: `a[j] = a[j-1]` — **한 방향으로만 복사**. 원래 값은 이미 `tmp`에 안전하게 빼놨으니 덮어써도 괜찮음

**종료 조건 2가지** (드모르간 법칙):
1. 왼쪽 끝 도달 (`j == 0`)
2. `tmp`보다 작거나 같은 원소 발견 (`a[j-1] <= tmp`)

→ **계속 조건**: `j > 0 and a[j-1] > tmp`

### 4. 삽입 정렬은 안정적이다 ✅
같은 값을 만나면 `a[j-1] > tmp`가 **거짓**(등호 없음)이라 **밀지 않고 멈춰**. 그래서 같은 값끼리 순서가 유지돼.

**장점**: 이미 정렬됐거나 거의 정렬된 배열에서 **매우 빠름**
**단점**: 삽입 위치가 멀면 **이동 횟수가 많아짐**

### 5. 이진 삽입 정렬 (binary insertion sort)
단순 삽입 정렬은 삽입 위치를 **선형으로** 찾아. 이걸 **7일차 이진 검색**으로 바꾸자! 🎯

```python
pl, pr = 0, i - 1
while True:
    pc = (pl + pr) // 2
    if a[pc] == key: break
    elif a[pc] < key: pl = pc + 1
    else: pr = pc - 1
    if pl > pr: break
```

**주의**: 개선된 건 **"위치 찾기"(O(n) → O(log n))**뿐이야. 위치를 찾은 뒤 **원소를 미는 비용은 여전히 O(n)**이라 전체 복잡도는 그대로 O(n²).

**간편 버전** (실습 6C-2): `bisect.insort(a, a.pop(i), 0, i)` — 표준 라이브러리가 다 해줌

### 6. 단순 정렬 3형제 (전부 O(n²))

| | 핵심 동작 | 안정성 | 강점 |
|---|---|---|---|
| **버블** (18일차) | 교환 | ✅ | 구현이 단순 |
| **선택** (오늘) | 선택 | ❌ | **교환 횟수 최소** |
| **삽입** (오늘) | 삽입 | ✅ | **거의 정렬된 데이터에 최강** |

> 교재 243p: 셋 다 O(n²)이라 효율이 좋지 않음 → 06-5절 셸 정렬부터 개선 시작

### 2. 🟢 [손] 선택 정렬 손으로 추적

*(①~⑥ 채우고 아래로 확인)*

````
def trace_selection(a):
    a = list(a)
    n = len(a)
    for i in range(n - 1):
        m = i
        for j in range(i + 1, n):
            if a[j] < a[m]:
                m = j
        print(f'i={i}: 최솟값 {a[m]}(인덱스{m}) ↔ a[{i}]={a[i]} 교환')
        a[i], a[m] = a[m], a[i]
        print(f'      → {a}')
    return a

trace_selection([6, 4, 8, 3, 1, 9, 7])
````

### 4. 🟢 [손] 삽입 정렬 손으로 추적 — "구멍" 따라가기

배열 `[6, 4, 1, 7, 3, 9, 8]` (교재 240p)

**핵심**: `tmp`에 값을 빼두면 그 자리가 **빈 구멍**이 되고, 구멍이 왼쪽으로 이동하다가 멈춘 자리에 `tmp`를 꽂아.

| i | 손에 쥔 값(tmp) | 삽입 위치 | 결과 |
|---|---|---|---|
| 1 | 4 | 인덱스 0 | `4 6 1 7 3 9 8` |
| 2 | 1 | **①____** | **②____** |
| 3 | 7 | **③____** | **④____** |

*(힌트: 7은 왼쪽 이웃(6)이 더 작으니 while이 아예 안 돌아 — 이동 0번!)*

````
def trace_insertion(a):
    a = list(a)
    n = len(a)
    print(f'시작: {a}')
    for i in range(1, n):
        j = i
        tmp = a[i]
        moves = 0
        while j > 0 and a[j-1] > tmp:
            a[j] = a[j-1]
            j -= 1
            moves += 1
        a[j] = tmp
        tag = f'{tmp} → 인덱스{j}에 삽입 (이동 {moves}번)' if moves else f'{tmp}는 이미 제자리 (이동 0번)'
        print(f'i={i}: {tag}')
        print(f'      → {a}')
    return a

trace_insertion([6, 4, 1, 7, 3, 9, 8])
````

### 7. 🟡 [설명] `a[j] = a[j-1]`은 왜 스왑이 아닐까 🔥

18일차 버블 정렬과 오늘 삽입 정렬의 코드를 나란히 봐:

```python
# 버블 정렬 — 스왑 (서로 맞바꿈)
a[j-1], a[j] = a[j], a[j-1]

# 삽입 정렬 — 단방향 대입 (한쪽만 이동)
a[j] = a[j-1]
```

- 삽입 정렬에서 `a[j] = a[j-1]`을 실행한 **직후** 배열을 보면 같은 값이 **두 자리**에 나타나. 왜 그래도 괜찮을까?

### 11. 🔴 [실험] 이진 삽입이 정말 더 빠를까

"위치 탐색 비교 횟수"만 따로 세서 단순 삽입과 비교해봐.

````
def insertion_search_compares(a):
    '''단순 삽입 정렬의 위치 탐색 비교 횟수'''
    n = len(a); total = 0
    for i in range(1, n):
        j = i; tmp = a[i]; cnt = 0
        while j > 0 and a[j-1] > tmp:
            cnt += 1; a[j] = a[j-1]; j -= 1
        if j > 0: cnt += 1
        a[j] = tmp
        total += cnt
    return total

def binary_search_compares(a):
    '''이진 삽입 정렬의 위치 탐색 비교 횟수'''
    n = len(a); total = 0
    for i in range(1, n):
        key = a[i]; pl, pr = 0, i - 1; cnt = 0
        while True:
            pc = (pl + pr) // 2
            cnt += 1
            if a[pc] == key: break
            elif a[pc] < key: pl = pc + 1
            else: pr = pc - 1
            if pl > pr: break
        total += cnt
        pd = pc + 1 if pl <= pr else pr + 1
        for j in range(i, pd, -1): a[j] = a[j-1]
        a[pd] = key
    return total

import random
print(f'{"n":>5} {"단순삽입 탐색비교":>16} {"이진삽입 탐색비교":>16} {"배율":>6}')
print('-' * 48)
for n in [10, 50, 100, 500, 1000]:
    x = list(range(n)); random.shuffle(x)
    c1 = insertion_search_compares(x[:])
    c2 = binary_search_compares(x[:])
    print(f'{n:>5} {c1:>16,} {c2:>16,} {c1/c2:>6.1f}x')
````

## ✅ 정답 & 해설

### R-1. 버블 vs 선택
- **버블**: 한 패스에서 **여러 번** 교환 가능 (이웃 쌍을 훑으며 조건 맞을 때마다 즉시 교환)
- **선택**: 한 패스에서 **정확히 1번** 교환 (다 훑어서 최솟값 위치만 기억 → 마지막에 한 번)
- **이유**: 버블은 "비교 즉시 교환" 전략, 선택은 "훑고 나서 한 번에 교환" 전략. 이 차이 때문에 선택 정렬의 교환 횟수가 훨씬 적어.

---
### R-2. 안정성 복습
- **안정적** = 값이 같은 원소들의 **원래 순서가 정렬 후에도 유지**되는 것
- **이웃끼리만 교환하면 안정적인 이유**: 같은 값 A, B가 이웃했을 때 `a[j-1] > a[j]`가 거짓이라 **교환이 일어나지 않아**. 그러니 A가 B를 앞지를 방법이 없어.
- **멀리 떨어진 원소를 교환하면**: 중간에 있는 같은 값을 **건너뛰어** 자리를 옮길 수 있어서 순서가 깨질 수 있음 → 오늘 6번에서 실증!

### 1. `min = i` 이름 가리기 🔥
**(a)** `min`은 함수 안의 **지역변수**야. 함수 내부에서만 그 이름이 정수로 덮이고, 정렬 로직(`a[j] < a[min]`)은 그 지역변수를 정상적으로 참조하니까 정렬은 잘 돼. 함수 밖의 내장 `min`은 영향 안 받아.

**(b)** `broken` 안에서는 정렬이 끝난 뒤에도 `min`이 여전히 **정수**를 가리켜. 거기서 `min(a)`처럼 **호출**하려 하니 `'int' object is not callable`.

**(c)** `min_idx`, `smallest`, `m` 같은 이름이 안전해. 교재가 `min`을 쓴 건 설명 편의상이고, 실무에선 피하는 게 좋아. (내장 이름 목록은 `dir(__builtins__)`로 확인 가능)

### 6. 선택 정렬 불안정성 🔥🔥
**(a)** **선택 정렬만** 뒤바뀌어:
- 버블: `(3,'L'), (3,'R')` — 유지 ✅
- 삽입: `(3,'L'), (3,'R')` — 유지 ✅
- **선택: `(3,'R'), (3,'L')` — 뒤바뀜** ❌

**(b)** 첫 패스(`i=0`)에서 최솟값 `1`(인덱스4)을 찾아 `a[0]`(`3ᴸ`)과 교환해. 그러면 **`3ᴸ`이 인덱스 4로 날아가.** 그런데 `3ᴿ`은 인덱스 3에 그대로 있으니, 이 순간 **`3ᴿ`이 `3ᴸ`보다 앞서게 돼.** 이후로는 이 역전이 회복되지 않아. 이게 "멀리 떨어진 원소 교환"이 안정성을 깨는 정확한 메커니즘이야.

**(c)** `a[j-1] > tmp`에 **등호가 없어서**, 같은 값을 만나면 조건이 **거짓**이 되어 밀기를 **멈춰**. 그래서 새로 삽입되는 원소는 항상 같은 값의 **뒤에** 자리 잡아 → 원래 순서 유지. (만약 `>=`로 바꾸면 같은 값도 밀어내서 안정성이 깨져 — 18일차 14번과 같은 원리)

---
### 7. `a[j] = a[j-1]`은 스왑이 아니다
**결과**: 실행 직후 같은 값이 두 자리에 나타나 (예: `[1,4,6,7,7,9,8]`).

**왜 괜찮나**: 원래 그 자리에 있던 값(`3`)은 이미 `tmp`에 **안전하게 백업**해뒀으니까. 배열 안에서 덮어써져도 진짜 값은 `tmp`에 살아있어. 나중에 `a[j] = tmp`로 되돌려놓으면 끝.

**tmp 없이 했다면**: `a[j] = a[j-1]` 하는 순간 원래 `a[j]`에 있던 값이 **영영 사라져.** 복구할 방법이 없어서 데이터가 손실돼. `tmp`는 "잃어버리지 않기 위한 임시 보관함"이야.

---
### 8. `j > 0`이 먼저인 이유
- `a[j-1] > tmp`를 먼저 검사하면, `j == 0`일 때 `a[-1]`(**배열의 맨 마지막 원소**)을 참조해버려 — 완전히 엉뚱한 값과 비교하게 됨.
- 파이썬 `and`는 **왼쪽이 거짓이면 오른쪽을 아예 평가 안 해**(단축 평가). `j > 0`이 거짓이면 `a[j-1]`은 실행조차 안 되니 사고가 원천 차단돼.
- **11일차 스택**: `peek`에서 `if self.is_empty(): raise Empty`처럼 **"경계를 먼저 확인하고, 안전할 때만 데이터 접근"**하는 패턴과 정확히 같아. 삽입 정렬 3번 문제(`if not stack`)도 같은 결.

### 9. 3형제 성능 비교
**(a) 역순에서 선택 3번 vs 버블·삽입 21번**: 선택 정렬은 **패스당 교환 1번**이 상한이야(`n-1=6`번이 최대, 실제 3번). 버블은 이웃 쌍마다 교환하고, 삽입은 매번 끝까지 밀어야 해서 각각 21번.

**(b) 이미 정렬됨에서 삽입만 6번 비교**: 삽입 정렬은 각 원소마다 **왼쪽 이웃 하나만 확인하고 바로 멈춰**(`a[j-1] > tmp`가 즉시 거짓). 그래서 원소당 비교 1번, 총 `n-1 = 6`번. **O(n²)이 O(n)으로 떨어지는 최선의 케이스**야.

**(c)** 선택 정렬은 "최솟값을 찾으려면" **정렬 안 된 구간을 무조건 끝까지 다 봐야** 해. 중간에 "이미 정렬됐네"를 알아챌 방법이 구조적으로 없어. 버블은 "이번 패스에 교환이 있었나"로 판단할 수 있었지만, 선택 정렬엔 그런 신호가 없어.

### 13. 종합
① **삽입 정렬** — 이미 정렬된 부분은 while이 거의 안 돌아 O(n)에 가까움
② **선택 정렬** — 교환이 최대 n-1번으로 가장 적음
③ **버블 또는 삽입** — 둘 다 안정적 (선택은 ❌)
④ **셋 다 부적합** — 전부 O(n²). n=10000이면 약 5천만 번 비교. 06-5절 **셸 정렬**, 이후 **퀵·병합 정렬**(O(n log n))이 필요

---

## 📌 핵심 3줄 요약

1. **선택 정렬**은 "최솟값 위치를 기억했다가 마지막에 한 번 교환" → **교환 횟수 최소(n-1번)**지만, 멀리 떨어진 원소를 교환해서 **불안정**하다.
2. **삽입 정렬**은 "값을 손에 쥐고(tmp), 이웃을 한 칸씩 밀며 자리를 만들어 꽂기" → **스왑이 아니라 단방향 대입**이고, **안정적**이며 **거의 정렬된 데이터에 최강**이다.
3. **이진 삽입 정렬**은 위치 탐색만 O(log n)으로 개선 — **원소 미는 비용 O(n)은 그대로**라 전체 복잡도는 여전히 O(n²).
