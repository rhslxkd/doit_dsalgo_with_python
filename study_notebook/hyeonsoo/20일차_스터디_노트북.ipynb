## 📦 부록 — 정렬 함수 모음

````
from typing import MutableSequence
import random

# ---------- 19일차: 단순 삽입 정렬 ----------
def insertion_sort(a, count=False):
    n = len(a); moves = 0
    for i in range(1, n):
        j = i
        tmp = a[i]
        while j > 0 and a[j-1] > tmp:
            a[j] = a[j-1]
            moves += 1
            j -= 1
        a[j] = tmp
    return moves if count else a

# ---------- 실습 6-8: 셸 정렬 (h //= 2) ----------
def shell_sort1(a, count=False):
    n = len(a); moves = 0
    h = n // 2
    while h > 0:
        for i in range(h, n):
            j = i - h
            tmp = a[i]
            while j >= 0 and a[j] > tmp:
                a[j+h] = a[j]
                moves += 1
                j -= h
            a[j+h] = tmp
        h //= 2
    return moves if count else a

# ---------- 실습 6-9: 셸 정렬 (h*3+1 수열) ----------
def shell_sort2(a, count=False):
    n = len(a); moves = 0
    h = 1
    while h < n // 9:
        h = h * 3 + 1
    while h > 0:
        for i in range(h, n):
            j = i - h
            tmp = a[i]
            while j >= 0 and a[j] > tmp:
                a[j+h] = a[j]
                moves += 1
                j -= h
            a[j+h] = tmp
        h //= 3
    return moves if count else a


# ---------- 임의의 h 수열로 셸 정렬 (실험용) ----------
def shell_with_h(a, h_list, count=False):
    n = len(a); moves = 0
    for h in h_list:
        for i in range(h, n):
            j = i - h
            tmp = a[i]
            while j >= 0 and a[j] > tmp:
                a[j+h] = a[j]
                moves += 1
                j -= h
            a[j+h] = tmp
    return moves if count else a

def h_seq_half(n):
    '''실습 6-8의 h 수열'''
    hs = []; h = n // 2
    while h > 0:
        hs.append(h); h //= 2
    return hs

def h_seq_3n1(n):
    '''실습 6-9의 h 수열'''
    h = 1
    while h < n // 9:
        h = h * 3 + 1
    hs = []
    while h > 0:
        hs.append(h); h //= 3
    return hs


def test_sort(f, trials=500):
    fails, example = 0, None
    for _ in range(trials):
        x = [random.randint(0, 50) for _ in range(random.randint(0, 15))]
        expected = sorted(x)
        y = x[:]
        f(y)
        if y != expected:
            fails += 1
            if example is None:
                example = (x, y, expected)
    return fails, example


def trace_h_sort(a, h):
    '''h-정렬 과정을 그룹별로 보여줌'''
    a = list(a)
    n = len(a)
    print(f'=== h={h} 정렬 ===')
    print(f'시작: {a}')
    for start in range(h):
        group_idx = list(range(start, n, h))
        group_val = [a[i] for i in group_idx]
        print(f'  그룹(인덱스 {group_idx}): {group_val}')
    for i in range(h, n):
        j = i - h
        tmp = a[i]
        while j >= 0 and a[j] > tmp:
            a[j+h] = a[j]
            j -= h
        a[j+h] = tmp
    print(f'결과: {a}')
    return a

print('준비 완료')
````

## 📖 오늘의 핵심 개념

### 1. 왜 셸 정렬인가 — 삽입 정렬의 한계
배열 `[1, 2, 3, 4, 5, 0, 6]`을 삽입 정렬하면:
- `2, 3, 4, 5`까지는 이미 정렬돼 있어서 **이동 0번** (아주 빠름 ✅)
- 그런데 `0` 하나를 맨 앞으로 보내려면 **5번 이동** (느림 ❌)

**근본 원인**: 삽입 정렬은 원소를 **한 칸씩만** 움직일 수 있어. 멀리 가려면 그만큼 여러 번 밀어야 해.

### 2. 셸 정렬(shell sort)의 아이디어 🎯
> **"멀리 떨어진 원소끼리 먼저 대충 정렬해두고, 간격을 좁혀가며 마무리하자"**

도널드 L. 셸(Donald L. Shell)이 고안. 퀵 정렬 이전까지 **가장 빠른 알고리즘**으로 알려졌었어.

**h-정렬**: `h`칸씩 떨어진 원소들끼리 묶어서 삽입 정렬하는 것

교재 248~249p 예시 (`[8,1,4,2,7,6,3,5]`, n=8):
```
h=4: (8,7) (1,6) (4,3) (2,5) 4개 그룹을 각각 정렬 → 7 1 3 2 8 6 4 5
h=2: (7,3,8,4) (1,2,6,5) 2개 그룹을 각각 정렬     → 3 1 4 2 7 5 8 6
h=1: 전체를 삽입 정렬                              → 1 2 3 4 5 6 7 8 ✓
```

**핵심**: `h`가 클 때는 **원소가 멀리 점프**할 수 있어서 큰 값/작은 값이 빠르게 제자리 근처로 가. 그 다음 `h`를 줄이면 이미 "거의 정렬된" 상태라 **삽입 정렬의 장점(빠름)**이 발휘돼!

### 3. 코드 구조 — 삽입 정렬과 뭐가 다른가

```python
# 19일차 삽입 정렬              # 오늘 셸 정렬
for i in range(1, n):           for i in range(h, n):      # 1 → h
    j = i                           j = i - h              # i → i-h
    tmp = a[i]                      tmp = a[i]
    while j > 0 and a[j-1] > tmp:   while j >= 0 and a[j] > tmp:   # j>0 → j>=0
        a[j] = a[j-1]                   a[j+h] = a[j]      # 1칸 → h칸
        j -= 1                          j -= h             # 1 → h
    a[j] = tmp                      a[j+h] = tmp
```

**"1"이 전부 "h"로 바뀐 것**뿐이야! `h=1`이면 정확히 삽입 정렬과 같아져.

### 4. ⚠️ h값 선택이 왜 중요한가 (교재 252p)
`h = 4 → 2 → 1`처럼 **h들이 서로 배수 관계**면 문제가 생겨.

교재 그림 6-18: `4-정렬`로 만든 그룹 `(8,7),(1,6),(4,3),(2,5)`를 `2-정렬`할 때, 파란 그룹 `((8,7),(4,3))`과 검은 그룹 `((1,6),(2,5))`이 **서로 섞이지 않아.** 결국 처음 상태와 비슷해져서 **애써 그룹으로 나눈 의미가 없어져.**

**해결**: h값이 서로 배수가 되지 않게 → `h = ... → 121 → 40 → 13 → 4 → 1` (`h*3+1` 수열)

```python
h = 1
while h < n // 9:      # 초깃값이 너무 크면 효과 없으므로 n//9 제한
    h = h * 3 + 1
...
h //= 3
```

> 💡 이거 어디서 본 것 같지 않아? **8~9일차 해시 테이블 크기를 소수로 잡았던 이유**와 같은 결이야 — "규칙적인 배수 관계는 데이터를 고르게 섞지 못한다"

### 5. 복잡도와 안정성
- **시간 복잡도**: **O(n^1.25)** — 단순 정렬 O(n²)보다 훨씬 빠름
- **안정성**: ❌ **안정적이지 않음** — 이웃하지 않고 **떨어져 있는 원소를 교환**하니까 (19일차 선택 정렬이 불안정했던 이유와 정확히 같은 원리)

### 6. 퀵 정렬 맛보기 (06-6 도입)
- **가장 빠른 정렬**로 널리 알려짐. 찰스 A. R. 호어가 고안
- **피벗(pivot)**: 그룹을 나누는 기준값
- 아이디어: 피벗보다 **작은 그룹 / 큰 그룹**으로 나누고, 각 그룹에서 또 피벗을 뽑아 나누기를 **재귀**로 반복
- 15일차 하노이·16일차 8퀸에서 배운 **분할 정복**이 여기서 또 나와! (다음 시간에 자세히)

## ✅ 정답 & 해설

### R-1. 삽입 정렬의 장점과 단점
- `[1,2,3,4,5,0,6]`에서 `0`을 맨 앞으로 보내려면 **5번** 이동 (`5,4,3,2,1`을 각각 한 칸씩 밀어야 함)
- 앞부분이 아무리 잘 정렬돼 있어도, `0`이 지나가야 할 **길목에 있는 원소들을 전부 한 칸씩 밀어야** 하기 때문
- **근본 원인**: 삽입 정렬은 원소를 **한 번에 한 칸씩만** 움직일 수 있어. 멀리 가려면 거리만큼 이동 횟수가 필요해.

### R-2. 안정성 총정리
① **이웃한 원소만 교환**하므로 같은 값끼리 순서가 안 바뀜
② **멀리 떨어진 원소를 교환**해서 중간의 같은 값을 건너뛸 수 있음
③ 같은 값을 만나면 `a[j-1] > tmp`가 거짓이라 **밀지 않고 멈춤**
④ **❌ 불안정** — 셸 정렬도 h칸 떨어진 원소를 교환하니까 (6번에서 실증)

### 1. 삽입 정렬의 한계
**(a)** `0` 때문에 **5번** 이동 (전체 이동도 5회)
**(b)** 삽입 정렬은 원소를 **한 칸씩만** 옮겨. 그래서 "제자리까지의 거리"가 곧 "이동 횟수"가 돼. 거의 정렬된 배열이라도 **딱 하나가 멀리 있으면** 그 하나 때문에 비용이 확 올라가.

### 2. 4-정렬
① `(7, 8)` / ② `(1, 6)` / ③ `(3, 4)` / ④ `(2, 5)`
⑤ **`7 1 3 2 8 6 4 5`**

교재 248p 그림 6-15와 일치. 아직 완전히 정렬되진 않았지만 **큰 값(8)이 뒤로, 작은 값(3)이 앞으로** 크게 점프한 걸 볼 수 있어.

### 3. 2-정렬 → 1-정렬
- 2-정렬 후: `3 1 4 2 7 5 8 6` (교재 249p 그림 6-16)
- 1-정렬 후: `1 2 3 4 5 6 7 8` ✓

**(a)** h를 **줄여가는** 이유: h가 클 때 원소를 멀리 점프시켜 **대략적인 자리**를 잡고, h가 작아질수록 **정밀하게** 다듬어. 반대로 `1 → 2 → 4`로 하면, 마지막 4-정렬이 4칸 떨어진 원소만 보므로 **전체 정렬이 보장되지 않아**(1-정렬로 끝나야 모든 이웃 쌍이 정렬됨).

**(b)** 19일차에 배운 **"삽입 정렬은 거의 정렬된 데이터에 아주 빠르다"**는 장점을 이용한 거야. h가 큰 단계들이 배열을 "거의 정렬된" 상태로 만들어두면, 마지막 1-정렬(=삽입 정렬)이 순식간에 끝나.

### 4. 셸 정렬 구현 정답
```python
h = n // 2
while h > 0:
    for i in range(h, n):
        j = i - h
        tmp = a[i]
        while j >= 0 and a[j] > tmp:
            a[j + h] = a[j]
            j -= h
        a[j + h] = tmp
    h //= 2
```
19일차 삽입 정렬에서 **"1"을 전부 "h"로** 바꾸고, `j > 0`을 `j >= 0`으로 바꾼 것뿐! (`j`가 `i-h`부터 시작하므로 0도 유효한 인덱스)

---
### 5. 셸 정렬의 속도
**결과**: n=500에서 삽입 62,558회 vs 셸 3,071회 — **약 20배** 차이

**(a)** 삽입 정렬은 **O(n²)**, 셸 정렬은 **O(n^1.25)**야. n이 커질수록 n²는 n^1.25보다 훨씬 빠르게 증가하니 격차가 벌어져. n이 10배 되면 삽입은 100배, 셸은 약 18배만 늘어.

**(b)** ① **약 5,600** (1000^1.25 ≈ 5,623)
O(n²)의 1,000,000과 비교하면 **약 178배** 적어. 같은 "느린 정렬" 계열이지만 실용성에서 큰 차이.

---
### 6. 셸 정렬의 불안정성 🔥🔥
**(a)** 입력 `[(2,'L'), (2,'R'), (1,'X'), (2,'M')]`:
- 삽입 정렬: `(1,X), (2,L), (2,R), (2,M)` — 순서 유지 ✅
- **셸 정렬: `(1,X), (2,R), (2,L), (2,M)`** — `2L`과 `2R`이 **뒤바뀜** ❌

**(b)** 셸 정렬은 `h`칸 떨어진 원소를 교환해. 예를 들어 `h=2`일 때 인덱스 0과 2를 바꾸면, 그 사이의 인덱스 1은 **건너뛰어져.** 만약 인덱스 1에 같은 값이 있었다면 순서가 뒤집혀. **19일차 선택 정렬**도 "멀리 떨어진 원소 교환" 때문에 불안정했는데, 완전히 같은 원리야.

**(c)** `h=1` 단계는 확실히 안정적이야. 하지만 그 **이전 단계(h>1)에서 이미 순서가 깨진 상태**로 들어오면, 안정적인 삽입 정렬은 "지금 이 순서"를 유지할 뿐 **원래 순서를 복원해주지 않아.** 한 번 깨진 안정성은 되돌릴 수 없어.

### 7. h값 배수 문제 🔥🔥
**(a)** `h=4`의 그룹0은 인덱스 `{0,4}`, 그룹2는 `{2,6}`이야. `h=2`의 그룹0은 인덱스 `{0,2,4,6}` — **4의 그룹0과 그룹2를 합친 것과 정확히 같아.** 즉 4-정렬로 나눈 그룹들이 **2-정렬에서도 여전히 같은 편**에 남아서, 서로 다른 그룹끼리 **비교될 기회가 없어.**

**4가 2의 배수**이기 때문에 이런 "그룹이 통째로 포함되는" 현상이 일어나.

**(b)** 그룹을 나눈 목적은 "**멀리 떨어진 원소들을 섞어서 대략적인 순서를 잡는 것**"이야. 그런데 그룹이 섞이지 않으면 **파란 그룹은 파란 그룹끼리, 검은 그룹은 검은 그룹끼리**만 정렬돼서, 합쳤을 때 처음 상태와 비슷해져 — **h를 여러 단계로 나눈 노력이 헛수고**가 되는 거야.

**(c)** **8~9일차 해시**에서 테이블 크기를 소수로 잡았던 이유와 정확히 같아:
- 해시: 키가 `6의 배수`처럼 규칙적이고 테이블 크기가 `12`(공약수 6)면 → 몇몇 버킷에만 몰림
- 셸: h들이 `4, 2, 1`처럼 배수 관계면 → 그룹이 섞이지 않고 몇몇 그룹에만 갇힘

**공통 원리**: **"규칙적인 배수 관계는 데이터를 고르게 섞지 못한다."** 서로소(공약수가 1뿐)인 관계를 만들어야 골고루 분산돼.

### 8. h*3+1 수열

...

**(b)** n=8이면 `8//9 = 0`이라 while이 한 번도 안 돌아서 `h=1`. 그럼 셸 정렬이 아니라 **그냥 단순 삽입 정렬**이 실행돼. (교재 254p 각주도 이걸 지적해)

### 9. 두 h 수열 비교 🔥

**측정 결과**:
- n=50~1000(2의 거듭제곱 아님): **`h//=2`가 더 적게 이동**
- n=128(2의 거듭제곱): **`h*3+1`이 더 적게 이동** (617 vs 721)

**(a)** `n=100`이면 h 수열은 `[50, 25, 12, 6, 3, 1]`이야. 잘 봐 — `50→25`는 배수지만 `25→12`는 **버림(`//`) 때문에 배수 관계가 깨져!** `25`와 `12`는 서로소야. 즉 **정수 나눗셈의 버림이 우연히 배수 관계를 끊어줘서** 7번의 문제가 완화돼.

**(b)** 맞아. `n=128`이면 `128//2=64, 64//2=32, ...`로 **버림이 한 번도 안 일어나** 순수한 2의 거듭제곱 `64,32,16,8,4,2,1`이 나와. 이때는 7번에서 본 "그룹이 안 섞이는" 문제가 그대로 발생해서 `h*3+1`이 이겨.

**(c)** 교재 주장은 **원리적으로는 맞아.** "h들이 서로 배수면 그룹이 안 섞여서 비효율적"은 사실이야(n=128에서 실증됨).

다만 **실제 `h //= 2` 구현에서는** 정수 나눗셈의 버림 덕분에 배수 관계가 자주 깨져서, 많은 n에서는 문제가 크게 드러나지 않아. **최악의 경우(n이 2의 거듭제곱)를 방어**한다는 점에서 `h*3+1`이 더 안전한 선택이고, 교재가 이걸 권하는 이유야.

> 💡 이런 게 "교과서 이론과 실제 측정을 둘 다 해봐야 하는 이유"야. 이론이 틀린 게 아니라, **이론이 말하는 조건이 언제 성립하는지**를 정확히 아는 게 중요해.

- **피벗(pivot)**: 그룹을 나누는 **기준값**. 교재 그림 6-19의 첫 피벗은 **학생 A의 키 168cm**
- 피벗 기준으로 **2개 그룹**(피벗 미만 / 피벗 이상)으로 나눔
- 나눈 각 그룹에서 **또 피벗을 뽑아 나누기를 반복**. **모든 그룹이 1명씩 남으면** 정렬 완료
- **공통점**: **분할 정복법(divide and conquer)** — 15일차 하노이("원반 n개 → n-1개 그룹 + 1개"), 16일차 8퀸("n열 배치 → 1열 배치 + 나머지 열")과 같은 전략. 큰 문제를 **같은 모양의 작은 문제**로 쪼개고 **재귀**로 해결!

---
### 11. 종합 정리
① **양방향 스캔** (홀수 패스는 뒤→앞, 짝수 패스는 앞→뒤)
② **최솟값을 선택**해서 맨 앞과 교환 (교환 횟수 최소)
③ 정렬된 부분의 **알맞은 위치에 삽입**
④ **O(n^1.25)**
⑤ **❌ 불안정**
⑥ **h칸 떨어진 원소끼리 먼저 정렬**하고 h를 줄여가며 마무리 (삽입 정렬의 개선판)

---

## 📌 핵심 3줄 요약

1. **셸 정렬**은 삽입 정렬의 단점("한 칸씩만 이동")을 개선한 것 — **h칸 떨어진 원소끼리 먼저 대충 정렬**해두고 h를 줄여가며, 마지막 `h=1`(=삽입 정렬)이 "거의 정렬된 배열"을 빠르게 마무리한다.
2. **h값은 서로 배수가 되지 않아야** 그룹이 골고루 섞인다 (8~9일차 해시 테이블 크기가 소수여야 했던 것과 같은 원리). 그래서 `h*3+1` 수열(`1→4→13→40→121`)을 쓴다.
3. 셸 정렬은 **O(n^1.25)**로 단순 정렬 O(n²)보다 훨씬 빠르지만, **떨어진 원소를 교환**하므로 **안정적이지 않다**.
