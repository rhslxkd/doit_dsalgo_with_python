## 📦 부록 — hanoi.py

이 셀을 **먼저 실행**해. (`input()` 부분은 노트북용으로 뺐어)

````
import io, contextlib

def move(no: int, x: int, y: int) -> None:
    '''원반 no개를 x기둥에서 y기둥으로 옮김'''
    if no > 1:
        move(no - 1, x, 6 - x - y)

    print(f'원반 [{no}]을(를) {x}기둥에서 {y}기둥으로 옮깁니다')

    if no > 1:
        move(no - 1, 6 - x - y, y)


def capture(f, *args):
    '''함수 출력을 줄 리스트로 캡처하는 헬퍼'''
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        f(*args)
    s = buf.getvalue().strip()
    return s.split('\n') if s else []

print('준비 완료')
````

## 📖 오늘의 핵심 개념 (완전 정리)

### 1. 하노이의 탑이란?
기둥 3개를 이용해 쌓아 놓은 원반을 **최소 횟수로** 옮기는 문제.

**규칙 2가지**:
1. 원반은 **1개씩만** 옮길 수 있다.
2. **큰 원반을 작은 원반 위에 쌓을 수 없다.**

**용어**:
| 이름 | 뜻 |
|---|---|
| 시작 기둥 | 원반이 처음 놓인 기둥 (1기둥) |
| 중간 기둥 | 중간에 놓는 기둥 (2기둥) |
| 목표 기둥 | 최종 목적지 (3기둥) |

> 📜 프랑스 수학자 **에두아르 뤼카(Édouard Lucas)**가 1883년에 발명. "황금 원반 64개를 3개의 탑 사이에서 바꿔 옮기면 종말이 찾아온다"는 인도 신화에서 유래. (8번 문제에서 이게 언제일지 계산해볼 거야)

### 2. 핵심 아이디어 — "그룹"으로 묶어 생각하기 🎯

원반 n개를 옮기는 문제는 이렇게 **3단계**로 쪼갤 수 있어:

```
① 맨 아래 원반을 제외한 그룹(원반 1 ~ n-1)을  시작 기둥 → 중간 기둥
② 맨 아래 원반(원반 n)을                      시작 기둥 → 목표 기둥
③ 그룹(원반 1 ~ n-1)을                        중간 기둥 → 목표 기둥
```

**왜 이렇게 나누나?** 맨 아래 원반(가장 큰 것)을 목표 기둥으로 옮기려면, 그 위에 있는 것들이 **전부 다른 기둥(중간)에 치워져 있어야** 하니까.

### 3. ⚠️ 가장 중요한 오해 포인트

> **"그룹을 옮긴다"는 건 실제 동작이 아니야.**

원반은 규칙상 **1개씩만** 옮길 수 있어. 그러니 "그룹 n-1개를 통째로 옮긴다"는 건 **실행할 수 없는 명령**이야. 그건 사람이 문제를 쪼개서 생각하기 위한 **개념적 도구**일 뿐이야.

그럼 어떻게 실행하나? → **똑같은 함수를 다시 호출**해서, 그 그룹도 또 "(더 작은 그룹) + (원반 1개)"로 쪼개. 이걸 반복하면 결국 **원반 1개 옮기기**라는 실행 가능한 단위까지 내려가.

| | 그룹 개념 | 재귀 호출 |
|---|---|---|
| 역할 | 문제를 **어떻게 쪼갤지** 생각 | 쪼갠 걸 **실제로 실행** |
| 실행 가능? | ❌ (선언일 뿐) | ✅ (원반 1개까지 내려감) |

**13~14일차와의 연결**: `factorial(n)`을 짤 때 `factorial(n-1)`이 내부적으로 어떻게 계산되는지 **몰라도** `n * factorial(n-1)`이라고 짤 수 있었지? "재귀 호출이 알아서 해줄 거야"라고 믿는 것 — 이걸 **믿음의 도약(leap of faith)**이라고 해. 하노이도 똑같아.

### 4. 코드 구조

```python
def move(no: int, x: int, y: int) -> None:
    if no > 1:
        move(no - 1, x, 6 - x - y)     # ① 그룹을 중간 기둥으로

    print(f'원반 [{no}]을(를) {x}기둥에서 {y}기둥으로')   # ② 바닥 원반 1개 이동

    if no > 1:
        move(no - 1, 6 - x - y, y)     # ③ 그룹을 목표 기둥으로
```

**매개변수**:
| 이름 | 의미 |
|---|---|
| `no` | 옮겨야 할 원반의 **개수** (= 맨 아래 원반의 **번호**) |
| `x` | **시작** 기둥 번호 |
| `y` | **목표** 기둥 번호 |

### 5. `6 - x - y` 트릭

기둥 번호는 1, 2, 3이고 **합이 6**이야. 그러니:

```
중간 기둥 = 6 - 시작 - 목표
```

시작과 목표가 어디든 **나머지 하나**가 자동으로 나와. (`if`문 여러 개 쓸 필요 없음)

### 6. 이동 횟수

원반 n개를 옮기는 최소 이동 횟수는 **2ⁿ - 1**회. (7번에서 직접 확인)

### 5. 🟡 [구조] "그룹을 옮긴다"는 실제로 무슨 일이 일어나는가 🔥🔥

**오늘의 핵심 문제야.** 아래 코드는 "그룹을 옮긴다"는 **선언**과 "원반 1개를 옮긴다"는 **실제 동작**을 구분해서 보여줘.

**실행 전에 예측**: `move(3,1,3)`에서 ★(실제 동작)은 총 몇 줄 나올까? "그룹" 선언은 몇 줄?

````
def move_traced(no, x, y, depth=0):
    ind = '  ' * depth
    mid = 6 - x - y
    if no > 1:
        print(f'{ind}[선언] 그룹(원반1~{no-1})을 {x}→{mid}기둥으로 옮겨야 함  (실행 불가! 다시 쪼갬)')
        move_traced(no - 1, x, mid, depth + 1)
    print(f'{ind}★[실행] 원반 {no}번 → 딱 1개만 이동: {x}기둥 → {y}기둥')
    if no > 1:
        print(f'{ind}[선언] 그룹(원반1~{no-1})을 {mid}→{y}기둥으로 옮겨야 함  (실행 불가! 다시 쪼갬)')
        move_traced(no - 1, mid, y, depth + 1)

move_traced(3, 1, 3)
````

**(a)** ★(실제 실행) 줄에서 한 번에 옮기는 원반은 **항상 몇 개**야? 왜 그럴 수밖에 없지? (규칙 2가지 중 하나)

**(b)** "[선언] 그룹을 옮겨야 함"이라고 말한 직후에 무슨 일이 일어나? 그 그룹은 통째로 옮겨졌어, 아니면 또 쪼개졌어?

### 7. 🟡 [실험/빈칸] 이동 횟수 공식 찾기

**실행 전에** 빈칸을 채워봐:

| n | 이동 횟수 | 이전 항과의 관계 |

### 10. 🔴 [빈칸] 재귀 호출 트리 완성 (교재 203p 그림 5-12)

`move(3, 1, 3)`의 재귀 호출 트리를 채워봐. 들여쓰기 = 재귀 깊이.

```
move(3, 1, 3)
├─ ① move(2, 1, ____)          ← 그룹(1,2)를 1→중간
│   ├─ ② move(1, 1, ____)      ← 그룹(1)을 1→중간
│   │   └─ ★ 원반 1: 1기둥 → ____기둥
│   ├─ ★ 원반 2: 1기둥 → ____기둥
│   └─ ③ move(1, ____, 2)      ← 그룹(1)을 중간→2
│       └─ ★ 원반 1: ____기둥 → 2기둥
├─ ★ 원반 3: 1기둥 → ____기둥
└─ ④ move(2, ____, 3)          ← 그룹(1,2)를 중간→3
    ├─ ⑤ move(1, 2, ____)
    │   └─ ★ 원반 1: 2기둥 → ____기둥
    ├─ ★ 원반 2: 2기둥 → ____기둥
    └─ ⑥ move(1, ____, 3)
        └─ ★ 원반 1: ____기둥 → 3기둥
```

*(빈칸 전부 채우고 아래로 확인)*

````
def move_tree(no, x, y, depth=1):
    ind = '    ' * depth
    mid = 6 - x - y
    if no > 1:
        print(f'{ind}├─ move({no-1}, {x}, {mid})   ← 그룹을 {x}→{mid}')
        move_tree(no - 1, x, mid, depth + 1)
    print(f'{ind}├─ ★ 원반 {no}: {x}기둥 → {y}기둥')
    if no > 1:
        print(f'{ind}└─ move({no-1}, {mid}, {y})   ← 그룹을 {mid}→{y}')
        move_tree(no - 1, mid, y, depth + 1)

print('move(3, 1, 3)')
move_tree(3, 1, 3)
print()
print('★ 개수를 세어봐 -> 2^3 - 1 = 7개여야 함')
````

## ✅ 정답 & 해설

### R-1. 재귀 3대 요소
- ① **2번** (`move(no-1, ...)`이 앞뒤로 두 번)
- ② **트리** (호출이 두 갈래로 갈라짐)
- ③ **순수한(genuinely)** 재귀 — 14일차 193p 용어. 재귀 호출을 여러 번 하는 함수
- ④ **`no == 1`** (정확히는 `if no > 1`의 거짓 쪽. 원반 1개면 그냥 옮기고 끝, 재귀 안 함)

### R-2. `recur`과 `move`의 공통 구조
- **공통 구조**: `재귀 호출 → 출력 → 재귀 호출`. 둘 다 출력을 **가운데** 두고 앞뒤로 재귀를 호출해. 그래서 실행 흐름이 트리 모양이고, 출력 순서가 "왼쪽 서브트리 전부 → 자기 자신 → 오른쪽 서브트리 전부"가 돼.
- **뒤쪽 재귀 호출** = **꼬리 재귀**(tail recursion). 함수 맨 끝이라 뒤에 할 일이 없으니, 14일차 방식대로 `while` 루프 + 인자 갱신으로 **제거 가능**해. (앞쪽 재귀는 `no`값을 나중에 출력해야 해서 스택이 필요)
- **출력 개수**: `recur`은 피보나치였지만 `move`는 **2ⁿ - 1**개. (7번 참고)

### 5. 그룹의 정체 🔥🔥

**(a)** **항상 1개**야. 규칙 1번("원반은 1개씩만 옮길 수 있다") 때문에 물리적으로 그럴 수밖에 없어. ★ 줄이 총 7줄인데, 이게 곧 실제 이동 횟수(2³-1=7)야.

**(b)** 통째로 옮겨진 게 **아니라 또 쪼개졌어.** "[선언]" 바로 다음 줄을 보면 한 단계 더 들여쓰기된 채로 또 다른 [선언]이나 ★가 나와. 결국 **★(원반 1개)까지 내려가야만** 실제로 뭔가 일어나.

**(c)** **핵심 답**:
- **그룹으로 묶는 이유**: 복잡한 문제("n개를 어떻게 옮기지?")를 **더 작은 같은 문제 + 실행 가능한 한 단계**로 쪼개서 생각하기 위한 **개념적 도구**. 사람이 이해하기 위한 사고 틀이야.
- **다시 푸는(재귀 호출하는) 이유**: "그룹을 옮긴다"는 건 **실행 가능한 명령이 아니야.** 규칙상 원반은 1개씩만 옮길 수 있으니까, 그 그룹도 똑같은 방식으로 계속 쪼개서 **원반 1개 옮기기**라는 실행 가능한 단위까지 내려가야 해.

**한 줄 요약**: **그룹은 "어떻게 쪼갤지"에 대한 아이디어, 재귀 호출은 "그 아이디어를 실제로 실행 가능하게 만드는" 절차.**

> 💡 13일차 `factorial`에서 `factorial(n-1)`이 어떻게 계산되는지 몰라도 `n * factorial(n-1)`이라 짤 수 있었던 것과 같아. "재귀 호출이 알아서 해줄 것"이라 믿는 **믿음의 도약(leap of faith)**.

### 6. `[no]개를` vs `[no]을(를)` 🔥

**(a)** **"원반 번호 `no` 하나"**야. 근거는 **print 앞뒤의 재귀 호출**:
```python
if no > 1: move(no-1, ...)   # 원반 1~no-1은 여기서 이미 처리됨
print(...)                    # 그러니 여기 남은 건 원반 no 하나뿐
if no > 1: move(no-1, ...)   # 나머지도 여기서 처리
```
앞뒤 재귀가 나머지 `no-1`개를 전부 담당하니까, `print` 자리에서 실제로 옮기는 건 **맨 아래 원반 1개(번호 = no)**뿐이야.

**(b)** 독스트링의 `no개`는 **맞는 표현**이야. `move(no, x, y)` 함수 **전체**가 하는 일은 "원반 no개를 옮기는 것"이 맞으니까. 하지만 `print`의 `no`는 **그중 딱 한 개(맨 아래 원반의 번호)**를 가리켜. **같은 변수 `no`가 두 가지 의미로 읽힌다는 게 헷갈림의 원인**이야:
- 함수 전체 관점: `no` = 옮겨야 할 원반 **개수**
- print 시점 관점: `no` = 지금 옮기는 원반의 **번호**

(원반이 크기 순으로 1~n번이라, 맨 아래 원반의 번호가 곧 개수와 같아서 이런 이중성이 생겨)

**(c)** 교재가 `[{no}]을(를)`로 쓴 건 **"3개를 한 번에 옮겼다"는 오해를 막기 위해서**야. "원반 [3]개를 옮깁니다"라고 읽으면 규칙 1번(1개씩만)을 어긴 것처럼 보이거든. 5번에서 다룬 "그룹은 실제 동작이 아니다"와 정확히 같은 오해 지점이야.

> 실용 팁: `print(f'원반 {no}번을 ...')` 처럼 **"번"**을 붙이면 더 명확해져.

### 7. 이동 횟수 공식

**(a) 점화식**: **`f(n) = 2 · f(n-1) + 1`** (단, `f(1) = 1`)
- 3단계 분해 그대로: 그룹 옮기기 `f(n-1)` + 원반 1개 `1` + 그룹 옮기기 `f(n-1)`

**(b) 왜 2ⁿ - 1인가**: 양변에 1을 더해보면 `f(n) + 1 = 2·(f(n-1) + 1)`. 즉 `g(n) = f(n)+1`이라 두면 `g(n) = 2·g(n-1)`이고 `g(1) = 2`니까 `g(n) = 2ⁿ`. 따라서 **`f(n) = 2ⁿ - 1`**.
(직관적으로도: 매 단계마다 이전 문제를 **두 번** 풀어야 하니 2배씩 커짐)

**(c)** **하노이(2ⁿ)가 훨씬 빨리 커져.**
- `recur`: `recur(n-1)` + `recur(n-2)` — 두 번째 호출이 **한 단계 더 작아서** 증가율이 황금비 φ≈**1.618**
- `move`: `move(n-1)` + `move(n-1)` — **똑같은 크기를 두 번** 부르니 증가율이 정확히 **2**

즉 "n-1을 두 번" vs "n-1과 n-2를 한 번씩"의 차이야. n=64면 하노이는 약 1844경, 피보나치식은 그보다 훨씬 작아.

### 8. 원반 64개

**결과**: 18,446,744,073,709,551,615회 (약 1844경). 1초에 1개씩이면 약 **5849억 년** — 우주 나이(138억 년)의 **약 42배**! 1초에 100만 개를 옮겨도 58만 년.

**(a)** 정확히 **약 2배**. `f(n) = 2·f(n-1) + 1`이니까 원반 1개 늘 때마다 이전 작업을 **두 번** 반복해야 해. (+1은 맨 아래 원반 1회라 무시할 만큼 작음)

**(b)** 비교해보면:
| 복잡도 | n=64일 때 |
|---|---|
| O(1) — 해시 (8일차) | 1회 |
| O(log n) — 이진검색 (7일차) | 6회 |
| O(n) — 선형검색 (6일차) | 64회 |
| **O(2ⁿ) — 하노이** | **1844경 회** |

O(log n)은 n이 2배가 돼도 +1회만 늘지만, O(2ⁿ)은 n이 **+1**만 돼도 **2배**가 돼. 컴퓨터가 아무리 빨라져도 지수 시간 문제는 n이 조금만 커지면 **원리적으로 못 푸는** 영역에 들어가.

**(c)** **더 잘할 수 없어. 2ⁿ-1이 최소야.**
증명 아이디어: 맨 아래 원반 n을 목표 기둥으로 옮기려면, 그 위의 n-1개가 **반드시** 중간 기둥에 전부 모여 있어야 해(규칙 2 때문). 그러려면 최소 `f(n-1)`회. 그 다음 원반 n을 1회. 그 다음 n-1개를 다시 목표로 옮기려면 또 `f(n-1)`회. 어떤 방법을 써도 이 세 단계는 피할 수 없으니 `f(n) ≥ 2·f(n-1) + 1`. 우리 알고리즘이 정확히 이 값을 달성하므로 **최적**이야.

---
### 9. 기둥 시뮬레이션 정답
```python
disk = pegs[x].pop()        # 맨 위 원반을 꺼냄
pegs[y].append(disk)        # 맨 위에 올림
```

**(a)** `pop()`(꺼내기)과 `append()`(올리기). 이건 **11일차 스택**과 정확히 같아 — 기둥은 맨 위에서만 넣고 뺄 수 있으니 **LIFO 구조**야. 하노이의 탑은 사실 **스택 3개를 조작하는 문제**인 셈.

**(b)** 규칙 위반이 **한 번도 없어.** 알고리즘이 자동으로 규칙을 지키는 이유: `move(no, x, y)`를 호출할 때 **"원반 1~no만" 다룬다는 게 보장**되고, 앞의 재귀 호출이 원반 1~no-1을 **전부 다른 기둥으로 치운 뒤에야** 원반 no를 옮기거든. 즉 원반 no를 옮기는 시점에 그 위에는 아무것도 없고, 목적지에도 no보다 작은 원반이 없어. **분해 방식 자체에 규칙이 내장**된 거야.

### 10. 재귀 호출 트리 정답
```
move(3, 1, 3)
├─ ① move(2, 1, 2)              ← 중간 = 6-1-3 = 2
│   ├─ ② move(1, 1, 3)          ← 중간 = 6-1-2 = 3
│   │   └─ ★ 원반 1: 1기둥 → 3기둥
│   ├─ ★ 원반 2: 1기둥 → 2기둥
│   └─ ③ move(1, 3, 2)          ← 중간(3)에서 목표(2)로
│       └─ ★ 원반 1: 3기둥 → 2기둥
├─ ★ 원반 3: 1기둥 → 3기둥
└─ ④ move(2, 2, 3)              ← 중간(2)에서 목표(3)로
    ├─ ⑤ move(1, 2, 1)          ← 중간 = 6-2-3 = 1
    │   └─ ★ 원반 1: 2기둥 → 1기둥
    ├─ ★ 원반 2: 2기둥 → 3기둥
    └─ ⑥ move(1, 1, 3)
        └─ ★ 원반 1: 1기둥 → 3기둥
```
★가 총 **7개** = 2³-1 ✓. 출력 순서는 위에서 아래로 읽으면 그대로 교재 실행결과와 일치해.

## 📌 핵심 3줄 요약

1. 하노이의 탑은 n개 원반 옮기기를 **"맨 아래 제외 그룹 → 중간" + "맨 아래 1개 → 목표" + "그룹 → 목표"** 3단계로 쪼개는 문제다.
2. **"그룹을 옮긴다"는 실행 가능한 명령이 아니라 생각의 도구**다. 규칙상 원반은 1개씩만 옮길 수 있으니, 재귀 호출로 계속 쪼개서 **원반 1개**까지 내려가야 실제로 실행된다.
3. 이동 횟수는 `f(n) = 2·f(n-1) + 1` = **2ⁿ - 1**. 원반 1개가 늘면 **2배**가 되는 지수 시간이라, 64개면 우주 나이의 42배가 걸린다.

## 🗂️ 스터디 진행 가이드
- 🟢 (1~4번): 전원 필수. **전부 빈칸 문제**라 손으로 채우면서 감 잡기
- 🟡 (5~7번): 팀 목표선.
  - **5번이 오늘의 핵심** 🔥🔥 — "그룹으로 묶었는데 왜 다시 푸나"의 답
  - 6번은 `[no]개를` vs `[no]을(를)` — 말장난 같지만 5번과 직결된 개념 문제
  - 7번 점화식 `f(n) = 2f(n-1) + 1`
- 🔴 (8~10번): 도전. 8번(원반 64개 = 5849억 년), 9번(기둥 = 스택!), **10번 재귀 트리 빈칸**
- **금요일 코딩테스트 범위**: 🟢🟡 (1~7번)

### 🎓 오늘 회수된 개념들
- **13일차 기저 조건 / 믿음의 도약** → `no == 1`, "재귀가 알아서 해줄 것" (5번)

- **11일차 스택 LIFO** → 기둥이 곧 스택 (`pop`/`append`) (9번)
