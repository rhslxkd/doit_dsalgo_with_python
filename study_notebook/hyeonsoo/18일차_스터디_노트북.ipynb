### R-2. 🟡 [설명] 정렬 알고리즘의 안정성 (교재 219p)

**안정적인(stable) 정렬**이란 **값이 같은 원소의 순서가 정렬 후에도 유지되는 것**을 말해.

아래 데이터는 `(점수, 학번)`이야. 이미 학번 순으로 정렬돼 있어.
```
(3,'A'), (1,'B'), (3,'C'), (2,'D'), (1,'E')
```

- **점수** 기준으로 오름차순 정렬했을 때, 안정적이라면 `(1,'B')`와 `(1,'E')` 중 뭐가 먼저 와야 해?
- 안정적이지 않다면 어떤 결과가 나올 수 있어?
- 실무에서 안정성이 중요한 상황을 하나 생각해봐. (힌트: "이름순 정렬 후 부서순 정렬")

````
data = [(3,'A'), (1,'B'), (3,'C'), (2,'D'), (1,'E')]

def bubble_pairs(a):
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):
            if a[j-1][0] > a[j][0]:
                a[j-1], a[j] = a[j], a[j-1]
    return a

print('정렬 전 :', data)
print('버블정렬 :', bubble_pairs(data[:]))
print()
print('같은 점수끼리 순서를 봐:')
print('  점수 1 → B, E 순서 유지?')
print('  점수 3 → A, C 순서 유지?')
````

## 📦 부록 — 정렬 함수 모음

이 셀을 **먼저 실행**해. 오늘 만든 4가지 알고리즘이 전부 들어있고, **비교/교환 횟수를 세는** 버전이야.

````
from typing import MutableSequence
import random

# ---------- 실습 6-1: 버블 정렬 (기본) ----------
def bubble1(a, count=False):
    '''기본 버블 정렬'''
    c = s = 0
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):
            c += 1
            if a[j-1] > a[j]:
                s += 1
                a[j-1], a[j] = a[j], a[j-1]
    return (c, s) if count else a

# ---------- 실습 6-3: 개선 1 — 교환 횟수로 중단 ----------
def bubble2(a, count=False):
    '''패스에서 교환이 0번이면 중단'''
    c = s = 0
    n = len(a)
    for i in range(n - 1):
        exchng = 0
        for j in range(n - 1, i, -1):
            c += 1
            if a[j-1] > a[j]:
                s += 1
                exchng += 1
                a[j-1], a[j] = a[j], a[j-1]
        if exchng == 0:        # ← 안쪽 for가 끝난 뒤!
            break
    return (c, s) if count else a

# ---------- 실습 6-4: 개선 2 — 스캔 범위 제한 ----------
def bubble3(a, count=False):
    '''마지막 교환 위치 이전은 이미 정렬됨'''
    c = s = 0
    n = len(a)
    k = 0
    while k < n - 1:
        last = n - 1
        for j in range(n - 1, k, -1):
            c += 1
            if a[j-1] > a[j]:
                s += 1
                a[j-1], a[j] = a[j], a[j-1]
                last = j
        k = last
    return (c, s) if count else a

# ---------- 실습 6-5: 셰이커 정렬 ----------
def shaker(a, count=False):
    '''양방향 버블 정렬'''
    c = s = 0
    left = 0
    right = len(a) - 1
    last = right
    while left < right:
        for j in range(right, left, -1):      # 오른쪽 → 왼쪽
            c += 1
            if a[j-1] > a[j]:
                s += 1
                a[j-1], a[j] = a[j], a[j-1]
                last = j
        left = last
        for j in range(left, right):          # 왼쪽 → 오른쪽
            c += 1
            if a[j] > a[j+1]:
                s += 1
                a[j], a[j+1] = a[j+1], a[j]
                last = j
        right = last                          # ← 안쪽 for가 끝난 뒤!
    return (c, s) if count else a


def test_sort(f, trials=300):
    '''정렬 함수가 올바른지 랜덤 테스트'''
    fails = 0
    example = None
    for _ in range(trials):
        x = [random.randint(0, 50) for _ in range(random.randint(2, 10))]
        expected = sorted(x)
        y = x[:]
        f(y)
        if y != expected:
            fails += 1
            if example is None:
                example = (x, y, expected)
    return fails, example


def trace_pass(a, pass_num=1):
    '''한 패스의 비교·교환 과정을 출력'''
    a = list(a)
    n = len(a)
    i = pass_num - 1
    print(f'패스 {pass_num} 시작: {a}')
    for j in range(n - 1, i, -1):
        mark = 'O 교환' if a[j-1] > a[j] else 'X'
        before = list(a)
        if a[j-1] > a[j]:
            a[j-1], a[j] = a[j], a[j-1]
        print(f'  j={j}: a[{j-1}]={before[j-1]} vs a[{j}]={before[j]}  →  {mark:6} {a}')
    print(f'패스 {pass_num} 종료: {a}')
    return a

print('준비 완료 — trace_pass, bubble1/2/3, shaker, test_sort() 사용 가능')
````

## 📖 오늘의 핵심 개념

### 1. 정렬이란? (06-1)
**키(key)** 값의 대소 관계에 따라 데이터를 일정한 순서로 늘어놓는 작업.
- **오름차순(ascending)**: 작은 값이 앞
- **내림차순(descending)**: 큰 값이 앞

### 2. 정렬 알고리즘의 안정성
**안정적(stable)** = 값이 같은 원소의 **원래 순서가 정렬 후에도 유지**됨.

버블 정렬은 **1칸 이상 떨어진 원소를 교환하지 않고 이웃한 원소만 교환**하므로 **안정적**이야. (교재 224p)

### 3. 내부 정렬 vs 외부 정렬
| | 설명 |
|---|---|
| **내부 정렬** | 데이터를 **하나의 배열에 다 담을 수 있을 때** |
| **외부 정렬** | 데이터가 너무 많아 하나의 배열에 못 담을 때 (별도 작업 파일 필요) |

이 책에서 다루는 8가지는 **전부 내부 정렬**이야.

### 4. 정렬의 3대 핵심 동작
> **교환 · 선택 · 삽입** (교재 220p)

대부분의 정렬 알고리즘이 이 셋을 응용해. 버블 정렬은 **교환**이 핵심.

### 5. 버블 정렬(bubble sort) = 단순 교환 정렬
**이웃한 두 원소**의 대소 관계를 비교해서 필요하면 **교환**하는 것을 반복.

**패스(pass)**: 배열 끝에서 앞쪽으로 훑으며 비교·교환하는 **한 번의 과정**.
- 패스 1번을 마치면 **가장 작은 원소가 맨 앞**으로 이동
- 패스를 k번 수행하면 **앞쪽 k개**가 정렬 완료
- 전체 정렬에는 **n-1번**의 패스가 필요

> **왜 n번이 아니라 n-1번?** — n-1개가 정렬되면 마지막 하나는 **자동으로** 제자리니까. (교재 222p)

**코드 구조**:
```python
for i in range(n - 1):                # 패스 (n-1번)
    for j in range(n - 1, i, -1):     # 뒤에서 앞으로 스캔
        if a[j-1] > a[j]:
            a[j-1], a[j] = a[j], a[j-1]
```
- `j`의 시작값은 `n-1` (맨 끝), 종룟값은 `i+1`
- 앞쪽 `i`개는 이미 정렬 완료라 건드리지 않음

**비교 횟수**: `(n-1) + (n-2) + ... + 1 = n(n-1)/2` → **O(n²)**

### 6. 개선 1 — 교환 횟수에 따른 중단 (실습 6-3)
어떤 패스에서 **교환이 한 번도 없었다면** = 이미 다 정렬된 것 → 이후 패스는 불필요.

```python
for i in range(n - 1):
    exchng = 0                        # 패스마다 0으로 초기화
    for j in range(n - 1, i, -1):
        if a[j-1] > a[j]:
            a[j-1], a[j] = a[j], a[j-1]
            exchng += 1
    if exchng == 0:                   # ★ 안쪽 for가 끝난 뒤에 검사!
        break
```

> ⚠️ **`if exchng == 0`의 위치가 생명이야.** 안쪽 `for` 루프 **안**에 들어가면 완전히 다른 의미가 돼. (9번 문제)

### 7. 개선 2 — 스캔 범위 제한 (실습 6-4)
어떤 패스에서 **마지막으로 교환한 위치보다 앞쪽**은 이미 정렬이 끝난 상태야. 그러니 다음 패스는 거기서부터 시작하면 돼.

```python
k = 0
while k < n - 1:
    last = n - 1                      # 이 패스의 마지막 교환 위치
    for j in range(n - 1, k, -1):
        if a[j-1] > a[j]:
            a[j-1], a[j] = a[j], a[j-1]
            last = j                  # 교환할 때마다 갱신
    k = last                          # 다음 패스의 시작점
```

`last`에는 **교환한 두 원소 중 오른쪽 원소(`a[j]`)의 인덱스**를 저장해.

### 8. 셰이커 정렬(shaker sort) = 양방향 버블 정렬
**문제**: `[9, 1, 3, 4, 6, 7, 8]`처럼 **큰 값이 맨 앞에** 있으면, 버블 정렬은 그 값을 **한 패스에 한 칸씩만** 뒤로 보내서 느려.

**해결**: 홀수 패스는 뒤→앞(작은 값을 앞으로), 짝수 패스는 앞→뒤(큰 값을 뒤로). **스캔 방향을 번갈아** 바꾸는 거야.

```python
left, right = 0, len(a) - 1
last = right
while left < right:
    for j in range(right, left, -1):   # ← 뒤에서 앞으로
        ...
        last = j
    left = last
    for j in range(left, right):       # → 앞에서 뒤로
        ...
        last = j
    right = last
```

**별명**: 양방향 버블 정렬, 칵테일 정렬, 칵테일 셰이커 정렬

### 2. 🟢 [빈칸] 비교 횟수 공식

버블 정렬의 총 비교 횟수를 채워봐. (원소 n개)

| 패스 | 비교 횟수 |
|---|---|
| 1번째 | n - 1 |
| 2번째 | **①____** |
| 3번째 | **②____** |
| ... | ... |
| (n-1)번째 | **③____** |
| **합계** | **④____** (공식) |

*(①~④ 채우고 아래로 확인)*

````
def count_per_pass(n):
    counts = []
    for i in range(n - 1):
        cnt = 0
        for j in range(n - 1, i, -1):
            cnt += 1
        counts.append(cnt)
    return counts

for n in [5, 7, 10]:
    c = count_per_pass(n)
    print(f'n={n:2}: 패스별 비교 {c}  합계={sum(c)}   n(n-1)/2={n*(n-1)//2}')
````

**질문**:
- "마지막 교환 위치보다 앞쪽은 이미 정렬됐다"가 **왜** 성립해?
- 만약 패스에서 교환이 **한 번도** 없었다면 `last`는 어떤 값이고, 그때 `while k < n-1` 조건은 어떻게 돼?

*(여기에 답 작성 → 아래로 확인)*

````
def trace_last(a):
    '''개선 2의 last / k 변화를 추적'''
    a = list(a)
    n = len(a)
    k = 0
    p = 1
    while k < n - 1:
        last = n - 1
        print(f'패스 {p}: k={k}, 스캔 범위 j = {n-1} → {k+1}')
        for j in range(n - 1, k, -1):
            if a[j-1] > a[j]:
                a[j-1], a[j] = a[j], a[j-1]
                last = j
        print(f'   결과 {a},  마지막 교환 위치 last={last}')
        k = last
        print(f'   → 다음 패스는 k={k}부터 (앞의 {k}개는 정렬 완료)')
        p += 1
    return a

trace_last([1, 3, 9, 4, 7, 8, 6])
````

```python

    exchng = 0                            exchng = 0
    for j in range(n-1, i, -1):           for j in range(n-1, i, -1):

            ...교환...                            ...교환...

        if exchng == 0:   ← for 안!       if exchng == 0:   ← for 밖
```

**실행 전에 예측**: 어느 쪽이 망가질까? 왜?

````
def bubbleA(a):
    '''if exchng == 0 이 안쪽 for 루프 안에 있는 버전'''
    n = len(a)
    for i in range(n - 1):
        exchng = 0
        for j in range(n - 1, i, -1):
            if a[j-1] > a[j]:
                a[j-1], a[j] = a[j], a[j-1]
                exchng += 1
            if exchng == 0:      # ← 안쪽 for 안!
                break
    return a

def bubbleB(a):
    '''if exchng == 0 이 안쪽 for 루프 밖에 있는 버전 (교재)'''
    n = len(a)
    for i in range(n - 1):
        exchng = 0
        for j in range(n - 1, i, -1):
            if a[j-1] > a[j]:
                a[j-1], a[j] = a[j], a[j-1]
                exchng += 1
        if exchng == 0:          # ← for 밖 (교재 실습 6-3)
            break
    return a

print('=== 교재 예시 [6,4,3,7,1,9,8] ===')
for name, f in [('버전 A (for 안)', bubbleA), ('버전 B (for 밖)', bubbleB)]:
    x = [6,4,3,7,1,9,8]
    f(x)
    ok = 'OK' if x == sorted(x) else '<<< 정렬 실패!'
    print(f'  {name:18}: {x}  {ok}')

print()
print('=== 랜덤 300회 테스트 ===')
for name, f in [('버전 A (for 안)', bubbleA), ('버전 B (for 밖)', bubbleB)]:
    fails, ex = test_sort(f, 300)
    print(f'  {name:18}: 실패 {fails}/300')
    if ex:
        print(f'      실패 예: {ex[0]} → {ex[1]}  (기대 {ex[2]})')
````

### 12. 🟡 [디버깅] `right = last`의 위치 🔥

`shaker_sort.py`에서 `right = last`가 **안쪽 for 루프 안**에 들어가 있었어.

```python
for j in range(left, right):
    if a[j] > a[j+1]:
        a[j], a[j+1] = a[j+1], a[j]
        last = j
    right = last        # ← for 안! (교재는 for 밖)
```

**9번의 `exchng` 버그와 달리, 이건 결과가 우연히 맞아.** 왜 그런지 확인해봐.

````
def shakerA(a, count=False):
    '''right = last 가 안쪽 for 안'''
    c = s = 0
    left, right = 0, len(a) - 1
    last = right
    while left < right:
        for j in range(right, left, -1):
            c += 1
            if a[j-1] > a[j]:
                s += 1; a[j-1], a[j] = a[j], a[j-1]; last = j
        left = last
        for j in range(left, right):
            c += 1
            if a[j] > a[j+1]:
                s += 1; a[j], a[j+1] = a[j+1], a[j]; last = j
            right = last          # ← for 안
    return (c, s) if count else a

print('=== 랜덤 300회 테스트 ===')
for name, f in [('셰이커 A (for 안)', shakerA), ('셰이커 B (for 밖, 교재)', shaker)]:
    fails, ex = test_sort(f, 300)
    print(f'  {name:24}: 실패 {fails}/300')

print()
print('=== 비교/교환 횟수도 같을까? ===')
for arr in [[9,1,3,4,6,7,8], [5,4,3,2,1], [3,1,4,1,5,9,2,6]]:
    a1 = shakerA(arr[:], count=True)
    a2 = shaker(arr[:], count=True)
    print(f'  {str(arr):26} A={a1}  B={a2}  {"동일" if a1==a2 else "다름!"}')
````

정말 그런지 직접 확인해봐.

````
def bubble_stable_check(a):
    '''(값, 태그) 튜플을 값 기준으로 정렬 — 태그 순서가 유지되는지 확인'''
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):
            if a[j-1][0] > a[j][0]:      # 값만 비교 (태그는 무시)
                a[j-1], a[j] = a[j], a[j-1]
    return a

def unstable_sort(a):
    '''일부러 안정성을 깬 버전: >= 로 비교 (같아도 교환)'''
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1, i, -1):
            if a[j-1][0] >= a[j][0]:     # ← >= 로 바꿈!
                a[j-1], a[j] = a[j], a[j-1]
    return a

data = [(2,'a'), (1,'b'), (2,'c'), (1,'d'), (2,'e')]
print(f'입력: {data}')
print(f'  버블(>)  : {bubble_stable_check(data[:])}')
print(f'  버블(>=) : {unstable_sort(data[:])}')
print()
print('같은 값의 태그 순서를 봐:')
print('  값 1 → b, d 순서가 유지되는가?')
print('  값 2 → a, c, e 순서가 유지되는가?')
````

- **안정적이면 `(1,'B')`가 먼저.** 원래 B가 E보다 앞에 있었으니까.
- 안정적이지 않으면 `(1,'E'), (1,'B')` 순서가 나올 수 있어.
- **실무 예**: 직원 목록을 **이름순**으로 정렬한 뒤 **부서순**으로 다시 정렬하면, 안정적인 정렬은 "부서별로 묶이되 각 부서 안에서는 이름순"이 유지돼. 불안정하면 부서 안의 이름 순서가 뒤죽박죽이 돼.

### 1. 패스가 n-1번인 이유
- 원소 7개 → 패스 **6번**
- **n-1번인 이유**: n-1개 원소의 정렬이 끝나면 **마지막 남은 원소는 자동으로 제자리**야. 비교할 상대가 없으니 패스를 한 번 더 할 필요가 없어. (교재 222p)
- 패스마다 비교 횟수: 1번째 n-1번, 2번째 n-2번, ... 매 패스마다 **1씩 감소**. 앞쪽 i개가 이미 정렬됐으니 건드릴 필요가 없거든.

---
### 2. 비교 횟수 공식
- ① **n - 2** / ② **n - 3** / ③ **1** / ④ **n(n-1)/2**

`(n-1) + (n-2) + ... + 2 + 1 = n(n-1)/2` (등차수열의 합)

---
### 3. 첫 번째 패스 정답
- ① **X** (1 < 8이라 교환 안 함) / ② `6 4 3 7 1 8 9`
- ③ **O** (7 > 1) / ④ `6 4 3 1 7 8 9`
- ⑤ `a[2]=3 vs a[3]=1` → O 교환 → `6 4 1 3 7 8 9`
- ⑥ `a[1]=4 vs a[2]=1` → O 교환 → `6 1 4 3 7 8 9`
- ⑦ `a[0]=6 vs a[1]=1` → O 교환 → `1 6 4 3 7 8 9`

**결과**: 맨 앞에 **1**이 왔어. 이건 배열 전체에서 **가장 작은 값**이야. 뒤에서부터 훑으며 작은 값을 계속 앞으로 밀어내니 자연스럽게 최솟값이 맨 앞에 도착해.

---
### 4. 두 번째 패스
- `j`는 `6`에서 시작해 `2`에서 끝나 (`range(6, 1, -1)`). 즉 인덱스 0은 건드리지 않아.
- 비교 횟수가 6번 → **5번**으로 1번 줄었어.
- 패스 종료 후 앞에서 **2개**(`1, 3`)가 정렬 완료.

### 5. 버블 정렬 구현 정답
```python
for i in range(n - 1):
    for j in range(n - 1, i, -1):
        if a[j-1] > a[j]:
            a[j-1], a[j] = a[j], a[j-1]
```

### 6. `exchng`가 감지하는 것
- 교환이 **0번**이라는 건, 스캔한 모든 이웃 쌍이 **이미 올바른 순서**였다는 뜻. 버블 정렬에서 "모든 이웃 쌍이 정렬됨" = **배열 전체가 정렬됨**이야.
- **패스마다 초기화하는 이유**: `exchng`는 "**이번 패스에서**" 교환이 있었는지를 봐야 해. 누적하면 이전 패스의 교환까지 섞여서 판단이 틀려.
- **효과 큰 입력**: 이미 정렬됐거나 거의 정렬된 배열 (첫 패스에서 바로 중단)
- **효과 없는 입력**: 역순 배열 (매 패스마다 교환이 일어나서 끝까지 감)

---
### 7. `last`가 저장하는 것
- **왜 앞쪽이 정렬됐나**: 마지막 교환 위치보다 앞쪽에서는 비교를 했지만 **교환이 한 번도 없었다**는 뜻이야. 버블 정렬은 이웃끼리 비교하니, 그 구간의 모든 이웃 쌍이 이미 올바른 순서 = 그 구간은 정렬 완료.
- **교환이 한 번도 없으면**: `last`는 초기값 `n-1` 그대로. 그럼 `k = n-1`이 되고 `while k < n-1` 조건이 **거짓**이 돼서 루프가 종료돼. 즉 **개선 1의 `exchng == 0` 중단이 자동으로 포함**돼 있는 셈이야! 이게 개선 2가 개선 1보다 강력한 이유.

### 8. 3버전 비교
**결과**: 기본 21/6, 개선1 20/6, 개선2 **12/6**

**(a)** 세 버전 모두 **같은 교환을 같은 순서로** 수행해. 개선은 "**불필요한 비교를 건너뛰는**" 것일 뿐, 교환 자체를 줄이는 게 아니야. 교환 횟수는 입력 배열이 정해지면 고정돼.

**(b)** 개선 1은 "**전체가 다 정렬됐을 때만**" 중단해 — 전부 아니면 전무. 개선 2는 "**앞쪽 일부가 정렬됐으면 그만큼 범위를 줄여**" — 매 패스마다 조금씩 이득을 봐. 게다가 7번에서 봤듯이 개선 2는 개선 1의 효과를 **자동으로 포함**해.

### 9. `if exchng == 0` 위치 🔥🔥🔥

**(a)** **버전 A(for 안)가 망가져.** 랜덤 300회 중 **200번 이상 실패**해.

**(b) 왜 망가지나**: 안쪽 for의 **첫 번째 비교**에서 교환이 안 일어나면 `exchng`가 **0인 채로** 바로 `if exchng == 0: break`를 만나. 그럼 **그 패스를 시작하자마자 중단**돼버려. 아직 앞쪽에 정렬 안 된 원소가 잔뜩 있는데도!
- 예: `[6,4,3,7,1,9,8]`의 두 번째 패스에서 첫 비교가 `8 vs 9`(교환 불필요) → 즉시 break → 나머지는 손도 못 댐

**(c)**
- `break`가 **안쪽 for**에 있으면 → **안쪽 for만** 탈출 (그 패스만 중단, 바깥 for는 계속)
- `break`가 **바깥 for**에 있으면 → **바깥 for 전체** 탈출 (정렬 종료)

파이썬에서 `break`는 **자기를 감싼 가장 가까운 반복문 하나만** 탈출해. 들여쓰기가 곧 "어느 반복문에 속하는가"를 결정하니, 한 칸 차이로 의미가 완전히 달라져.

**(d) 발견하기 어려운 이유**: **일부 입력에서는 우연히 정답이 나와.** 예를 들어 첫 비교부터 계속 교환이 일어나는 배열이면 `exchng`가 0이 아니라서 break를 안 타고 정상 동작해. 그래서 테스트를 몇 개만 해보면 "잘 되네?" 하고 넘어가기 쉬워.
→ **랜덤 테스트를 여러 번 돌려보는 게 중요한 이유**야. (13일차 `factorial(-1)`, 9일차 오픈해시 `None` 버그와 같은 **조용한 실패**)

### 10. 셰이커가 필요한 이유
**결과**: 버블 3종 전부 21/6, 셰이커는 **15/6**

**(a)** 9가 맨 뒤로 가려면 **6칸**을 이동해야 하는데, 버블 정렬의 한 패스는 9를 **딱 한 칸만** 뒤로 보내. (뒤→앞 스캔이라 큰 값은 밀려나기만 함) 그래서 패스 6번이 필요해.

**(b)** 반대로 **작은 값이 맨 뒤에** 있으면 **한 패스만에** 맨 앞까지 와. 뒤→앞으로 스캔하면서 작은 값이 계속 앞으로 끌려오거든. 즉 버블 정렬은 **작은 값에겐 빠르고 큰 값에겐 느려** — 비대칭이야.

> 별명 유래: 큰 값은 천천히 가라앉고(거북이, turtle), 작은 값은 빠르게 떠오름(토끼, rabbit).

**(c)** 셰이커는 **방향을 번갈아** 바꿔서 이 비대칭을 없애. 뒤→앞 패스는 작은 값을 앞으로, 앞→뒤 패스는 큰 값을 뒤로 보내니 **양쪽 다 빠르게** 제자리를 찾아가.

---
### 11. 셰이커 방향 추적
**(a)** 첫 패스(뒤→앞): 가장 작은 값 **1**이 맨 앞으로 왔어. `left = 1`이 돼서 인덱스 0은 확정.
**(b)** 두 번째 패스(앞→뒤): 가장 큰 값 **9**가 맨 뒤로 갔어. `right`가 줄어들어서 뒷부분도 확정.
**(c)**
- `left` = 스캔 범위의 **첫 원소 인덱스** (이보다 앞은 정렬 완료)
- `right` = 스캔 범위의 **마지막 원소 인덱스** (이보다 뒤는 정렬 완료)
- `left`와 `right`가 만나면(`left >= right`) 정렬 완료

### 12. `right = last` 위치 🔥

**(a) 왜 결과가 같나**: 두 가지 이유가 겹쳐:
1. `for j in range(left, right)`의 `range` 객체는 **루프가 시작될 때 이미 생성**돼. 루프 안에서 `right`를 바꿔도 **이번 루프의 반복 범위는 변하지 않아.**
2. `right`는 안쪽 for 내부에서 **다른 용도로 쓰이지 않아** (비교는 `a[j]`, `a[j+1]`만 사용).

그래서 매 반복마다 `right = last`를 해도, 루프가 끝난 시점의 `right` 값은 **마지막 `last` 값**과 같아 — 결국 for 밖에서 한 번 대입한 것과 동일한 결과.

**(b) 9번과의 차이**: 9번의 `if exchng == 0: break`는 **제어 흐름(break)**을 바꿔서 **실제 동작이 달라졌어.** 12번의 `right = last`는 **값 대입**일 뿐이고, 그 값이 루프 진행에 영향을 주지 않아서 **결과가 같아.**

→ 9번은 **진짜 버그**, 12번은 **우연히 무해한 스타일 문제**.

**(c) 그래도 for 밖이 나은 이유**:
1. **의도가 명확**: "이 패스가 끝나면 범위를 좁힌다"는 뜻인데, for 안에 있으면 "매 비교마다 범위를 바꾼다"로 읽혀서 오해를 부름
2. **불필요한 연산**: 루프 횟수만큼 대입이 반복됨 (성능 차이는 미미하지만 낭비)
3. **미래의 버그 위험**: 나중에 누가 루프 안에서 `right`를 사용하는 코드를 추가하면 **그때 진짜 버그가 터져**. (9일차 `[Bucket()] * capacity`가 "지금은 우연히 동작하지만 나중에 터지는" 것과 같은 종류)

### 13. 최선·최악·평균

**(a) 이미 정렬됨**: 기본 버블은 아무 교환도 없는데 **끝까지 n(n-1)/2번 비교**해. 개선 버전은 **첫 패스 끝나자마자 중단**(비교 n-1번)해서 O(n²) → **O(n)**이 돼. 극적인 차이!

**(b) 역순 배열**: 매 패스마다 교환이 계속 일어나서 `exchng`가 0이 안 되고, `last`도 계속 앞쪽으로 갱신돼. 즉 **줄일 범위 자체가 없어** → 어떤 개선도 무용지물, 전부 O(n²).

**(c)** `[10,1,2,...,9]`는 **큰 값 하나만 맨 앞**에 있는 최악의 버블 케이스야(10번 참고). 버블은 10을 한 칸씩 밀어야 해서 패스를 여러 번 돌지만, 셰이커는 **앞→뒤 패스 한 번에** 10을 맨 뒤로 보내버려.

### 14. 안정성 검증

**(a)** `>` 버전은 같은 값의 태그 순서가 **유지**되고(`b,d` / `a,c,e`), `>=` 버전은 **뒤집혀**(`d,b` / `e,c,a`).

**(b) 왜 깨지나**: `>=`는 **값이 같아도 교환**해. 값이 같은 두 원소 A, B가 이웃했을 때 굳이 자리를 바꾸니, 원래 순서가 파괴돼. `>`는 "**앞이 뒤보다 클 때만**" 교환하므로 같은 값끼리는 절대 자리를 안 바꿔 → 안정성 유지.

**(c) 실무 예**: 학생 목록을 **이름순**으로 정렬한 뒤 **학년순**으로 다시 정렬하면, 안정적인 정렬은 "학년별로 묶이되 각 학년 안에서는 이름순"이 유지돼. 불안정하면 두 번째 정렬이 첫 번째 정렬 결과를 망쳐버려서, 원하는 "다중 기준 정렬"을 못 해.

### 15. 내림차순 정답
```python
if a[j-1] < a[j]:      # 뒤→앞: 부등호 뒤집기
...
if a[j] < a[j+1]:      # 앞→뒤: 부등호 뒤집기
```
**부등호만 뒤집으면 돼.** `left`/`right`/`last` 로직은 그대로 — "정렬 기준"만 바뀌고 "탐색 전략"은 동일하니까.

### 16. O(n²)의 의미

**(a)** n이 10배 → 비교 횟수는 약 **100배** (n²이니까 10² = 100).
- n=10: 45회 → n=100: 4,950회 → n=1000: 499,500회

**(b)** ① **약 500,000** (정확히는 499,500)

O(log n)은 1000일 때 10인데 O(n²)은 50만 — **5만 배** 차이야.

**(c)** **바꿀 수 없어.** 개선 1·2와 셰이커는 **최선/평균 케이스**를 개선할 뿐, **최악(역순 배열)에서는 여전히 O(n²)**이야. 실험에서도 역순 배열은 네 버전 전부 `n(n-1)/2`번 비교했지.
→ 진짜로 복잡도를 낮추려면 **알고리즘 자체를 바꿔야** 해. 그게 앞으로 배울 **퀵 정렬·병합 정렬 O(n log n)**이야.

### 17. 시각화 정답
```python
mark = '  ' if m != j - 1 else (' +' if a[j-1] > a[j] else ' -')
```
- `m != j-1`: 지금 비교 중인 위치가 아니면 공백 2칸
- `m == j-1`이고 교환하면 `' +'`, 교환 안 하면 `' -'`

## 📌 핵심 3줄 요약

1. **버블 정렬**은 이웃한 두 원소를 비교·교환하는 **패스**를 n-1번 반복한다. 비교 횟수 `n(n-1)/2` → **O(n²)**, 이웃끼리만 교환하므로 **안정적**.
2. **개선 1**(교환 0이면 중단)과 **개선 2**(마지막 교환 위치로 범위 제한)는 **비교 횟수만** 줄인다. 교환 횟수는 그대로. 개선 2가 개선 1의 효과를 자동으로 포함한다.
3. **셰이커 정렬**은 스캔 방향을 번갈아 바꿔서, 버블 정렬의 "큰 값은 느리게 이동" 비대칭을 해소한다. 단 **최악의 경우는 여전히 O(n²)**.
